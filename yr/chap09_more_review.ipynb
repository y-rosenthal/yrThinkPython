### Question 1 (medium): write a function

Write a function called `split_sentences` that takes a list of lines from a book and returns a list of the book's sentences.

A sentence may start on one line and finish on a later line, and one line may contain more than one sentence. A sentence ends with a period, a question mark, or an exclamation point.

For example:

```python
split_sentences(["Hello world. How are",
                 "you today? I am",
                 "fine!"])
# ['Hello world.', 'How are you today?', 'I am fine!']
```

Rules, so there is no guessing:

- Join the lines first, with a single space between lines, so words split across lines do not run together, and so a sentence that spans lines comes back as one sentence. Ignore blank lines.
- A piece of text with no ending mark at the end is not a complete sentence, so leave it out.
- **Simplification — read this carefully:** for this question, assume that every period ends a sentence. Do NOT worry about abbreviations such as "Mr." or "Mrs.", or about acronyms such as "U.S.A." — in real books those periods do not end a sentence, but here you may treat every period as a sentence ending. That means `split_sentences(["Mr. Smith went home."])` returns `['Mr.', 'Smith went home.']`, and that is correct for this question.

Hint: the string method `split` makes a list from a string by splitting on a character — for example, `'a.b.c'.split('.')` gives `['a', 'b', 'c']`. You may split on a period, a question mark, or an exclamation point — but remember that `split` throws the character you split on away, so if you use it, you must put the ending mark back on each sentence, or you will lose that text. Splitting is not the only way to do this; the Answer shows two ways.

Solution 1 (mark the ends, then split):

```python
def split_sentences(lines):
    words = []
    for line in lines:
        stripped = line.strip()
        if stripped:
            words.append(stripped)
    text = ' '.join(words)
    for end in '.?!':
        text = text.replace(end, end + '\n')
    sentences = []
    for piece in text.split('\n'):
        piece = piece.strip()
        if piece and (piece.endswith('.') or piece.endswith('?') or piece.endswith('!')):
            sentences.append(piece)
    return sentences
```

After joining the lines, every ending mark gets a newline put right after it. Splitting on the newline then gives one piece per sentence, with its ending mark still attached. The last `if` drops a final piece with no ending mark.

Solution 2 (walk the text, one character at a time):

```python
def split_sentences(lines):
    words = []
    for line in lines:
        stripped = line.strip()
        if stripped:
            words.append(stripped)
    text = ' '.join(words)
    sentences = []
    current = ''
    for ch in text:
        current += ch
        if ch in '.?!':
            sentences.append(current.strip())
            current = ''
    return sentences
```

`current` collects one sentence. The `if` inside the `for` is the whole decision: when the character is an ending mark, the sentence is finished, so it is added to the list and `current` starts over. A leftover `current` with no ending mark is never added, which is exactly the rule above. Both solutions pass all the examples in the question.

```python
def sentences_from_file(filename):
    f = open(filename)
    lines = []
    for line in f:
        stripped = line.strip()
        if stripped:
            lines.append(stripped)
    return split_sentences(lines)
```

`sentences_from_file` does only the file part — opening the file and making a list of its lines — and hands the list to `split_sentences`, which does only the sentence part. If the sentence rules change, only Question 1's function changes.

```python
def count_questions(sentences):
    count = 0
    for sentence in sentences:
        if sentence.endswith('?'):
            count += 1
    return count
```

```python
def longest_sentence(sentences):
    longest = ''
    for sentence in sentences:
        if len(sentence) > len(longest):
            longest = sentence
    return longest
```

For example, `checkerboard_print(2, 3)` prints:

```
* *
 *
```

Hint: a character is `'*'` when its row plus its column is even.

```python
def checkerboard_print(rows, cols):
    for r in range(rows):
        line = ''
        for c in range(cols):
            if (r + c) % 2 == 0:
                line += '*'
            else:
                line += ' '
        print(line)
```

The inner loop builds one row, character by character, and the `if` inside the inner loop decides each character.

### Question 6 (hard, in two parts): write functions

A list of lists can hold a table: each inner list is one row, and each item in a row is one cell. For example:

```python
table = [['Name', 'Age', 'City'],
         ['Ada', '36', 'London'],
```

**Part A: first pass — one row per line.** Write a function called `print_grid` that prints a table like this, one row per line, with a single space between the items in a row.

For example, `print_grid(table)` prints:

```

Bo 7 New York
```

```python

    ...
```

<details>
<summary>Answer</summary>

```python
def print_grid(grid):
    for row in grid:
        cells = []
        for item in row:
            cells.append(str(item))
        print(' '.join(cells))
```

`str(item)` matters: a cell may be a number, and `join` needs strings. This version is easy to read, but the columns do not line up, because the items have different lengths — `'New York'` is longer than `'Bo'`, so the next column starts in a different place on each row.

**Part B: second pass — every item in the same width.** Modify Part A's function so that the columns line up. First work out the widest item in the whole table (the longest string, once every item has been turned into a string), then print every item padded to that width, with a single space between items. Call it `print_grid_aligned`.

For example, `print_grid_aligned(table)` prints (trailing spaces at the end of a row are harmless):

```

```

Now `'Age'` starts in the same column on every row, and so does `'City'`, because every item takes up 8 characters (the width of `'New York'`) plus the separating space.

```python
def print_grid_aligned(grid):               # CHANGED: new name, for the new job
    width = 0                               # ADDED: the widest item, so far
    for row in grid:                        # ADDED: first pass over the table
        for item in row:                    # ADDED: inner loop over the cells
            if len(str(item)) > width:     # ADDED: is this item wider?
                width = len(str(item))     # ADDED: then it is the new widest
    for row in grid:                        # UNCHANGED
        cells = []                          # UNCHANGED
        for item in row:                    # UNCHANGED
            cells.append(str(item).ljust(width))  # CHANGED: was str(item); now str(item).ljust(width)
        print(' '.join(cells))              # UNCHANGED
    # DELETED: nothing, except the idea that one pass is enough.
    # Part B needs two passes: one to find the width, one to print.
```

What did NOT change: the outer loop over the rows, joining a row's items with a single space, and turning each item into a string. What changed: the function name, and the item in the `join` (now padded). What was added: the whole first pair of nested loops that finds `width`. What was deleted: nothing from Part A's printing logic — it is still there, with padding added to it.

Why two passes are needed: you cannot pad an item to the widest width until you know what the widest width is, and you only know that after you have looked at every item. A version that pads each column to its own widest item is possible, and is even neater, but it is not asked for here — here every item gets the same width, the widest item in the whole table.

</details>
