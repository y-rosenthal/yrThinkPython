## What is a nested loop?

A loop inside another loop is called a **nested loop**. The outer loop runs as usual, but each
time it runs its body once, the inner loop runs all of its passes, from start to finish.

The indentation shows which loop a statement belongs to: statements indented once are in the
outer loop; statements indented twice are in the inner loop.

````python
for i in [1, 2]:
    print('outer loop, i =', i)
    for j in ['a', 'b']:
        print('   inner loop, j =', j)
````

The outer loop runs twice, with `i` equal to `1` and then `2`. Each time the outer loop runs
its body once, the inner loop runs twice, with `j` equal to `'a'` and then `'b'`. So the
inner `print` runs 4 times in total: 2 outer passes times 2 inner passes.

## Example 1: a multiplication table

Nested loops are the natural way to fill in a grid of values, with rows and columns. The outer
loop chooses the row, and the inner loop chooses the column.

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end=' ')
    print()
````

For each row `i`, the inner loop prints `i * 1`, `i * 2` and `i * 3` on one line (`end=' '`
puts a space after each product instead of a newline). Then `print()` with no arguments ends
the row and moves to the next line. The table has 3 rows because the outer loop runs 3 times.

## Example 2: the words in a file, and the letters in each word

Chapter 12 will loop through the lines of a book and, inside that loop, through the words in
each line. The same pattern works with the word list: the outer loop takes one line of the
file at a time, and the inner loop takes one letter of the word at a time.

````python
count = 0
for line in open('words.txt'):
    word = line.strip()
    print(word + ':')
    for letter in word:
        print('  ' + letter)
    count += 1
    if count == 2:
        break
````

## Example 3: every pair from two lists

To combine every element of one list with every element of another list, loop through the
first list and, inside that loop, loop through the second list.

````python
colors = ['red', 'blue']
shapes = ['circle', 'square']
for color in colors:
    for shape in shapes:
        print(color, shape)
````

## Example 4: counting with an inner loop

The counter pattern from Chapter 7 works inside a nested loop, too. Here the outer loop takes
one word at a time, and the inner loop counts the vowels in that word. Notice that `count`
is set back to `0` for each word: the counting starts over inside the outer loop, before the
inner loop runs.

````python
for word in ['apple', 'sky', 'hello']:
    count = 0
    for letter in word:
        if letter in 'aeiou':
            count += 1
    print(word, 'has', count, 'vowels')
````

<details>
<summary>Diving deeper: give the two loops different variable names</summary>

The outer and inner loops each have their own loop variable, and both variables exist at the
same time. If you use the same name for both, the inner loop overwrites the outer loop's
variable, which is confusing:

```python
for i in range(2):
    for i in range(3):
        print(i, end=' ')
    print('-> after the inner loop, i is', i)
```

The inner loop sets `i` to `0`, `1`, `2`, so when the inner loop finishes, `i` is `2`, not the
outer loop's value. (The outer loop still runs twice, because `range(2)` gives `i` a new value
at the start of each pass, but the code is needlessly hard to read.) Use different names for
the two loop variables, like `i` and `j`, or `word` and `letter`.

</details>

### Question 2 (easy): how many times?

How many times does the body of the inner loop run here?

```python
for i in range(4):
    for j in range(5):
        print('hi')
```

<details>
<summary>Answer</summary>

20 times. The outer loop runs 4 times, and the inner loop runs 5 times for each outer pass,
so 4 times 5 = 20. In general, multiply the number of outer passes by the number of inner
passes.

Write a function called `total_letters` that takes a list of words (strings) and returns the

```python
def total_letters(word_list):
    total = 0
    for word in word_list:
        for letter in word:
            total += 1
    return total
```

The outer loop takes one word at a time; the inner loop takes one letter of that word at a

### Question 6 (hard): find the bug

This loop is supposed to display the pairs `(0, 0)`, `(0, 1)`, `(1, 0)`, `(1, 1)`, but the
output is confusing. What is wrong, and how do you fix it?

```python
for i in range(2):
    for i in range(2):
        print(i, i)
```

<details>
<summary>Answer</summary>

Both loops use the same variable name, `i`, so the inner loop overwrites the outer loop's
variable. The fix is to give the inner loop its own name:

```python

    for j in range(2):
        print(i, j)
```

See "Diving deeper: give the two loops different variable names" above.
