## A reminder: how to call a method

Chapter 8 calls a method with the dot operator: `word.upper()`. The general form is:

```
STRING.METHOD(ARGUMENTS)
```

STRING is any string: a variable, a string in quotes, or an expression that makes a string.

````python
word = 'banana'
word.upper()
````

````python
'banana'.upper()
````

Strings are immutable (Chapter 8), so a string method never changes the string. It returns a **new**
string. After the call, `word` has the same value as before. (Only the value of the last line is displayed.)

````python
word = 'banana'
word.upper()
word
````

## Methods that take arguments: [`startswith()`](https://docs.python.org/3/library/stdtypes.html#str.startswith) and [`endswith()`](https://docs.python.org/3/library/stdtypes.html#str.endswith)

The methods above take no arguments. Many string methods take arguments, in the parentheses, as functions do.
Chapter 8 uses two of them: `startswith` and `endswith`.

**Optional arguments.** Some arguments are **optional**: you can give them, or leave them out. You saw this before: `input` works
with a prompt or without one (Chapter 5), and `open` works with the mode `'w'` or without it (Chapter 8).
When you leave out an optional argument, the method uses a **default value**.

On this page, the general form of a method shows each correct way to call it. The descriptions tell which
arguments are optional, and what their default values are.

**Quick summary.** Each line shows a call, and the value that it returns. The details of each method are
below.

````python
# @title Quick summary
date = '2026-10-21'
print('date                         ', repr(date))
print("date.startswith('2026')      ", repr(date.startswith('2026')))
print("date.startswith('10', 5)     ", repr(date.startswith('10', 5)))
print("date.startswith('10', 5, 7)  ", repr(date.startswith('10', 5, 7)))
print("date.endswith('21')          ", repr(date.endswith('21')))
print("date.endswith('21', 8)       ", repr(date.endswith('21', 8)))
print("date.endswith('10', 0, 7)    ", repr(date.endswith('10', 0, 7)))
````

### Details: startswith and endswith

<div class="yr-collapse">

To hide or show the details of each method, click the arrow next to this heading.

</div>

#### [`startswith()`](https://docs.python.org/3/library/stdtypes.html#str.startswith)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.startswith(PREFIX)
STRING.startswith(PREFIX, START)
STRING.startswith(PREFIX, START, END)
```

`startswith` returns `True` if STRING starts with PREFIX, and `False` if it does not.

- PREFIX: the characters to look for at the start.
- START (optional): the index where the check starts. The default is `0`, the start of the string.
- END (optional): the index where the check stops. As in a slice, the character at index END is not part of
  the check. The default is the end of the string.

With START and END, `startswith` checks the slice `STRING[START:END]`. For example,
`date.startswith('10', 5, 7)` gives the same result as `date[5:7].startswith('10')`.

</div>

The examples use a date, written year-month-day. Only PREFIX: does the date start with `'2026'`?

````python
date = '2026-10-21'
date.startswith('2026')
````

PREFIX and START: the month starts at index 5. Is the month October (`'10'`)?

````python
date = '2026-10-21'
date.startswith('10', 5)
````

PREFIX, START and END: check only the characters at indexes 5 and 6 (the month):

````python
date = '2026-10-21'
date.startswith('10', 5, 7)
````

#### [`endswith()`](https://docs.python.org/3/library/stdtypes.html#str.endswith)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.endswith(SUFFIX)
STRING.endswith(SUFFIX, START)
STRING.endswith(SUFFIX, START, END)
```

`endswith` returns `True` if STRING ends with SUFFIX, and `False` if it does not.

- SUFFIX: the characters to look for at the end.
- START (optional): the index where the check starts. The default is `0`, the start of the string.
- END (optional): the index where the check stops. As in a slice, the character at index END is not part of
  the check. The default is the end of the string.

With START and END, `endswith` checks the slice `STRING[START:END]`. For example,
`date.endswith('10', 0, 7)` gives the same result as `date[0:7].endswith('10')`.

</div>

Only SUFFIX: is it the 21st day of the month?

````python
date = '2026-10-21'
date.endswith('21')
````

SUFFIX and START: check from index 8 (the day) to the end:

````python
date = '2026-10-21'
date.endswith('21', 8)
````

SUFFIX, START and END: does the part from index 0 to index 6 (the year and the month) end with `'10'`?

````python
date = '2026-10-21'
date.endswith('10', 0, 7)
````

The comparison is case sensitive: `'A'` and `'a'` are different characters. To ignore case, call
[`lower()`](https://docs.python.org/3/library/stdtypes.html#str.lower) first. Then call `endswith` on the result: Python calls the methods from left to right.

````python
filename = 'Vacation.JPG'
print(filename.endswith('.jpg'))
print(filename.lower().endswith('.jpg'))
````

#### [`count()`](https://docs.python.org/3/library/stdtypes.html#str.count)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.count(SUB)
STRING.count(SUB, START)
STRING.count(SUB, START, END)
```

`count` returns the number of times that SUB is in STRING. (SUB means *substring*: a part of a string.)
Chapter 8 uses `count` to count the times that the name "Jonathan" is in *Dracula*.

- SUB: the string to count.
- START, END (optional): count only in the slice `STRING[START:END]`, as for `startswith`. The defaults are the

</div>

Only SUB:

````python
'banana'.count('a')
````

SUB can have more than one character:

````python
'banana'.count('an')
````

`count` does not count matches that overlap. `'ana'` is in `'banana'` at index 1 and at index 3, but these
two matches share the `a` at index 3, so `count` finds only the first one:

````python
'banana'.count('ana')
````

SUB and START: count from index 2 to the end (`'nana'`):

````python
'banana'.count('a', 2)
````

SUB, START and END: count from index 2 up to index 4, without index 4 (`'na'`):

````python
'banana'.count('a', 2, 4)
````

`count` is case sensitive. To count upper and lower case letters together, call `lower()` first:

````python
print('Banana'.count('b'))
print('Banana'.lower().count('b'))
````

#### [`strip()`](https://docs.python.org/3/library/stdtypes.html#str.strip)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.strip()
STRING.strip(CHARS)
```

`strip` returns a copy of STRING without the **whitespace** at its start and at its end. Whitespace is the
characters that you do not see: spaces, tabs, and newlines (`'\n'`). Whitespace in the middle of the string
stays.

- CHARS (optional): the characters to remove, in place of whitespace. CHARS is a *set of characters*, not a
  word: `strip` removes each of these characters, in any order, from the two ends.

Chapter 8 uses `line.strip()` to remove the newline at the end of each line of a file.

</div>

Without CHARS, `strip` removes the spaces at the two ends, but not the space in the middle:

````python
'   hello world   '.strip()
````

It also removes newlines:

````python
'hello\n'.strip()
````

With CHARS: remove the dots and the exclamation marks at the two ends:

````python
'...Wait!!!'.strip('.!')
````

Be careful: CHARS is a set of characters. `strip('ab')` removes each `a` and each `b` at the two ends, until
it finds a different character:

````python
'banana'.strip('ab')
````

Users often type a space before or after their answer. They do not see it, but the program does. Here, the
user typed a space before `yes`, so `likes_python` (from the section on user input) does not understand
the answer:

````python
def likes_python():
    answer = input('Do you like Python? Type yes or no: ')
    if answer.lower() == 'yes':
        print('Great! Me too.')
    elif answer.lower() == 'no':
        print('Maybe you will like it at the end of this course.')
    else:
        print('Please type yes or no.')

likes_python()
````

The fix is to call `strip()` and then `lower()` on the answer. Python calls the methods from left to right:
`strip()` returns a new string, then `lower()` is called on that string.

````python
def likes_python():
    answer = input('Do you like Python? Type yes or no: ')
    answer = answer.strip().lower()
    if answer == 'yes':
        print('Great! Me too.')
    elif answer == 'no':
        print('Maybe you will like it at the end of this course.')
    else:
        print('Please type yes or no.')
````

````python
likes_python()
````

#### [`lstrip()`](https://docs.python.org/3/library/stdtypes.html#str.lstrip) and [`rstrip()`](https://docs.python.org/3/library/stdtypes.html#str.rstrip)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.lstrip()
STRING.lstrip(CHARS)

STRING.rstrip()
STRING.rstrip(CHARS)
```

`lstrip` and `rstrip` are like `strip`, but they remove characters from one end only: `lstrip` from the
left end (the start), and `rstrip` from the right end (the end).

- CHARS (optional): the characters to remove, as for `strip`. The default is whitespace.

</div>

````python
'   hello   '.lstrip()
````

````python
'   hello   '.rstrip()
````

For example, a line of a file can start with spaces that you want to keep. `rstrip('\n')` removes only the
newline at the end:

````python
'    indented line\n'.rstrip('\n')
````

With CHARS: remove the dollar signs at the start only:

````python
'$$$42.50$'.lstrip('$')
````

#### [`removeprefix()`](https://docs.python.org/3/library/stdtypes.html#str.removeprefix) and [`removesuffix()`](https://docs.python.org/3/library/stdtypes.html#str.removesuffix)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.removeprefix(PREFIX)
STRING.removesuffix(SUFFIX)
```

`removeprefix` returns a copy of STRING without PREFIX at its start. `removesuffix` returns a copy of STRING
without SUFFIX at its end. If STRING does not start with PREFIX (or end with SUFFIX), they return STRING
unchanged.

- PREFIX, SUFFIX: the exact string to remove. Unlike CHARS for `strip`, it is a string, not a set of
  characters, and it is removed only once.

</div>

````python
'www.example.com'.removeprefix('www.')
````

````python
'www.example.com'.removesuffix('.com')
````

If the string does not start with PREFIX, nothing changes:

````python
'Smith'.removeprefix('Mr. ')
````

Compare with `strip`. `strip('w.com')` removes each of the characters `w`, `.`, `c`, `o` and `m` from the two
ends, so it also removes the `.com`:

````python
'www.example.com'.strip('w.com')
````

And compare with `lstrip`. `lstrip('a')` removes all the `a` characters at the start, but `removeprefix('a')`
removes only one:

````python
print('aaab'.lstrip('a'))
print('aaab'.removeprefix('a'))
````

## Lining up text: center, ljust and rjust

These methods add characters to a string, until the string has a given length. They are useful to show text
in columns, as in a table.

**Quick summary.** Each line shows a call, and the value that it returns. The details of each method are
below.

````python
# @title Quick summary
print("'Python'.center(12)       ", repr('Python'.center(12)))
print("'Python'.center(12, '*')  ", repr('Python'.center(12, '*')))
print("'Python'.ljust(12, '.')   ", repr('Python'.ljust(12, '.')))
print("'Python'.rjust(12, '.')   ", repr('Python'.rjust(12, '.')))
````

### Details: center, ljust and rjust

<div class="yr-collapse">

To hide or show the details of each method, click the arrow next to this heading.

#### [`center()`](https://docs.python.org/3/library/stdtypes.html#str.center)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.center(WIDTH)
STRING.center(WIDTH, FILLCHAR)
```

`center` returns a copy of STRING in the middle of a new string of length WIDTH, with characters added at the
two ends.

- WIDTH: the length of the new string. If STRING has WIDTH characters or more, `center` returns STRING
  unchanged: it never makes a string shorter.
- FILLCHAR (optional): the character to add. It must be exactly one character. The default is a space, `' '`.

</div>

````python
'Python'.center(20)
````

````python
'Python'.center(20, '*')
````

WIDTH is shorter than the string, so the string does not change:

````python
'Python'.center(3)
````

FILLCHAR must be one character:

````python
%%expect TypeError

'Python'.center(20, '*-')
````

#### [`ljust()`](https://docs.python.org/3/library/stdtypes.html#str.ljust) and [`rjust()`](https://docs.python.org/3/library/stdtypes.html#str.rjust)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.ljust(WIDTH)
STRING.ljust(WIDTH, FILLCHAR)

STRING.rjust(WIDTH)
STRING.rjust(WIDTH, FILLCHAR)
```

`ljust` returns a copy of STRING, *left justified* in a new string of length WIDTH: STRING is at the left, and
the characters are added at the right. `rjust` is the opposite (*right justified*): STRING is at the right,
and the characters are added at the left.

- WIDTH, FILLCHAR: the same as for `center`.

</div>

````python
'Python'.ljust(10, '.')
````

````python
'Python'.rjust(10, '.')
````

Here, `ljust` and `rjust` line up the items and the prices of a menu. The function `print_row` makes each
item 12 characters long and each price 7 characters long, so the columns line up. `str` (Chapter 2) changes
the price into a string, because `rjust` is a string method.

````python
def print_row(item, price):
    print(item.ljust(12, '.') + str(price).rjust(7))

print('MENU'.center(19, '='))
print_row('Coffee', 2.5)
print_row('Bagel', 3.25)
print_row('Orange juice', 4)
````

## Finding where a substring is: [`find()`](https://docs.python.org/3/library/stdtypes.html#str.find), [`rfind()`](https://docs.python.org/3/library/stdtypes.html#str.rfind), [`index()`](https://docs.python.org/3/library/stdtypes.html#str.index) and [`rindex()`](https://docs.python.org/3/library/stdtypes.html#str.rindex)

The `in` operator (Chapter 8) tells you **if** a string contains a substring. The methods in this section
tell you **where** it is: they return its index. Why do you need the index? With the index, you can slice the
string (Chapter 8) at that place, to get the part before it or the part after it.

`index` and `rindex` are like `find` and `rfind`, but when the substring is not there, they cause a
`ValueError` in place of returning `-1`.

**Quick summary.** Each line shows a call, and the value that it returns. The details of each method are
below.

````python
# @title Quick summary
print("'one fish two fish'.find('fish')    ", repr('one fish two fish'.find('fish')))
print("'one fish two fish'.rfind('fish')   ", repr('one fish two fish'.rfind('fish')))
print("'one fish two fish'.find('cat')     ", repr('one fish two fish'.find('cat')))
print("'one fish two fish'.rfind('cat')    ", repr('one fish two fish'.rfind('cat')))
print("'one fish two fish'.index('fish')   ", repr('one fish two fish'.index('fish')))
print("'one fish two fish'.rindex('fish')  ", repr('one fish two fish'.rindex('fish')))
````

### Details: find, rfind, index and rindex

#### [`find()`](https://docs.python.org/3/library/stdtypes.html#str.find)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.find(SUB)
STRING.find(SUB, START)
STRING.find(SUB, START, END)
```

`find` returns the index of the first time that SUB is in STRING (the index of the first character of SUB).
If SUB is not in STRING, `find` returns `-1`.

- SUB: the string to find.
- START, END (optional): look only in the slice `STRING[START:END]`, as for `count`. The index that `find`
  returns is an index of STRING, not of the slice.

</div>

For example, an email address has a user name before the `@` and a domain after it. `find` gives the index
of the `@`:

````python
email = 'mary.smith@example.com'
at = email.find('@')
at
````

Slices with this index give the two parts:

````python
email = 'mary.smith@example.com'
at = email.find('@')
print(email[:at])
print(email[at + 1:])
````

With START, `find` can find the second time that a substring is in a string. First, find the first time.
Then, look again, from the next index:

````python
text = 'one fish two fish red fish'
first = text.find('fish')
second = text.find('fish', first + 1)
print(first, second)
````

With START and END, look only from index 0 up to index 7 (`'one fis'`). The word `fish` is not complete
there, so `find` returns `-1`:

````python
text = 'one fish two fish red fish'
text.find('fish', 0, 7)
````

**Always check for `-1`.** `-1` is a correct index (the last character), so a slice with it does not cause an
error. It gives a wrong result, and you do not see the problem. Here, the address has no `@`, and the slice
removes the last letter:

````python
email = 'marysmith'
at = email.find('@')
email[:at]
````

This program asks for an email address and checks for `-1` before it slices:

````python
def ask_email():
    email = input('What is your email address? ')
    at = email.find('@')
    if at == -1:
        print('That is not an email address: it has no @.')
    else:
        print('User name:', email[:at])
        print('Domain:', email[at + 1:])
````

````python
ask_email()
````

````python
ask_email()
````

#### [`rfind()`](https://docs.python.org/3/library/stdtypes.html#str.rfind)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.rfind(SUB)
STRING.rfind(SUB, START)
STRING.rfind(SUB, START, END)
```

`rfind` is like `find`, but it returns the index of the **last** time that SUB is in STRING (*r* is for
*right*: it searches from the right end). If SUB is not in STRING, `rfind` returns `-1`.

- SUB, START, END: the same as for `find`.

</div>

For example, the extension of a file name is the part after the last dot. A file name can have more than one
dot, so `find` gives the wrong dot:

````python
filename = 'my.vacation.photo.jpg'
print(filename.find('.'))
print(filename.rfind('.'))
````

````python
filename = 'my.vacation.photo.jpg'
filename[filename.rfind('.') + 1:]
````

In the same way, the file name in a path is the part after the last `/`:

````python
path = 'C:/Users/mary/Documents/notes.txt'
path[path.rfind('/') + 1:]
````

#### [`index()`](https://docs.python.org/3/library/stdtypes.html#str.index) and [`rindex()`](https://docs.python.org/3/library/stdtypes.html#str.rindex)

<div class="yr-method" style="border-left: 6px solid #3b82f6; background: rgba(59,130,246,0.08); padding: 0.5em 1em; margin: 0.5em 0;">

```
STRING.index(SUB)
STRING.index(SUB, START)
STRING.index(SUB, START, END)

STRING.rindex(SUB)
STRING.rindex(SUB, START)
STRING.rindex(SUB, START, END)
```

`index` is like `find`, and `rindex` is like `rfind`, with one difference: if SUB is not in STRING, they
cause a `ValueError`, in place of returning `-1`.

- SUB, START, END: the same as for `find`.

</div>

````python
'mary.smith@example.com'.index('@')
````

````python
%%expect ValueError

'marysmith'.index('@')
````

Which one do you use?

- Use `find` (or `rfind`) when it is normal that the substring is not there, for example in text that a user
  types. Check for `-1`, as `ask_email` does.
- Use `index` (or `rindex`) when the substring must be there. If it is not there, there is a bug, and the
  error tells you about it at once. With `find`, the `-1` can give a wrong result that you do not see.
