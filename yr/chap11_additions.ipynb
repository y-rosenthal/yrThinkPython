## What is a while loop?

A `while` statement repeats a block of code while a condition is true. Its general form is:

```python
while CONDITION:
    BODY
```

`CONDITION` is an expression that is `True` or `False`, like the condition of an `if`
statement. `BODY` is the indented block after the colon. Python checks the condition before
each run of the body. If the condition is true, the body runs. If the condition is false, the
loop ends, and Python runs the code after the loop.

A `while` statement looks like an `if` statement, but it behaves differently after the body
runs. After the body of an `if` statement runs, Python moves on to the code after the `if`.
After the body of a `while` statement runs, Python checks the condition again. If it is still
true, the body runs again. The loop repeats while the condition stays true.

Here is a countdown:

````python
n = 3
while n > 0:
    print(n)
    n = n - 1
print('Blastoff!')
````

Here is what happens, step by step. Before the loop, `n` is `3`.

- Python checks the condition `n > 0`. It is true, so the body runs: it displays `3` and sets `n` to `2`.
- Python checks the condition again. It is true, so the body runs: it displays `2` and sets `n` to `1`.
- Python checks the condition again. It is true, so the body runs: it displays `1` and sets `n` to `0`.
- Python checks the condition again. `0 > 0` is false, so the loop ends. Python runs the code after the loop, which displays `Blastoff!`.

Each run of the body is one **iteration**. We say that the loop **iterates**: this loop iterates
3 times. To follow a loop, write the values of the variables on paper and update them line by
line, as above.

## When to use a while loop

Use a `while` loop when you cannot tell, at the time you write the code, how many times the loop will run. Use a `for` loop when you can tell: when the values are ready before the loop starts.

This `for` loop runs 5 times. The list is ready before the loop starts, so the count is known when you write the code:

````python
for n in [3, 1, 4, 1, 5]:
    print(n * 2)
````

This `while` loop doubles `n` until it passes 100. How many times does it run? You cannot tell by reading the code. You have to run it. The loop stops when the condition turns false, not after a set count:

````python
n = 1
while n <= 100:
    n = n * 2
print(n)
````

The rule is about what you know when you write the code. If the values are ready first, use a `for` loop. If the loop must keep going until something happens, use a `while` loop. The guessing game later on this page is an example: no one knows how many guesses it will take.

## The parts of a while loop

When you write a `while` loop, keep these rules in mind:

1. The condition depends on at least one variable. Here the condition is `n > 0`, and the variable is `n`.
2. The variable in the condition gets its value before the loop starts. Here `n = 3` comes before the `while` statement.
3. The body changes a variable in the condition, so that the condition can become false. Here `n = n - 1` makes `n` smaller each time, until `n > 0` is false.
4. The body is the indented block after the colon. Only the indented lines repeat.
5. The condition is checked before each iteration. If it is false at the first check, the body never runs:

````python
n = 0
while n > 0:
    print(n)
    n = n - 1
print('Done')
````

The condition `0 > 0` is false before the first check, so the body never runs, and only `Done` is displayed.

## Example: factorial

In math, the **factorial** of a whole number is the product of that number and all the
positive whole numbers below it. The factorial of 5 is `5 * 4 * 3 * 2 * 1`, which is 120.
Mathematicians write 5! for the factorial of 5. The factorial of 0 and the factorial of 1
are both defined to be 1.

Python's `math` module has a function that computes factorials:

````python
import math
print(math.factorial(0))
print(math.factorial(1))
print(math.factorial(5))
````

Here is the same function written with a `while` loop. It takes a whole number `n`, 0 or more:

1. `result = 1` sets up the answer before the loop starts.
2. The condition is `n > 1`. The loop repeats while `n` is more than 1.
3. The body multiplies `result` by `n`, then subtracts 1 from `n`.
4. When `n` is 1, the condition is false, the loop ends, and the function returns `result`.

````python
def factorial(n):
    result = 1
    while n > 1:
        result = result * n
        n = n - 1
    return result

print(factorial(0))
print(factorial(1))
print(factorial(3))
print(factorial(5))
````

For `factorial(3)`, the values change like this:

- At the start, `result` is `1` and `n` is `3`.
- First iteration: `3 > 1` is true, so `result` becomes `3` and `n` becomes `2`.
- Second iteration: `2 > 1` is true, so `result` becomes `6` and `n` becomes `1`.
- The check `1 > 1` is false, so the loop ends, and the function returns `6`.

For `factorial(0)` and `factorial(1)`, the condition is false at the first check, so the body
never runs, and the function returns `1`, which is the correct answer.

## Example: Fibonacci numbers

The **Fibonacci numbers** are the sequence 0, 1, 1, 2, 3, 5, 8, 13, ... The first two numbers
are 0 and 1. Every other number is the sum of the two numbers before it.

This function returns the first `n` Fibonacci numbers as a list. It uses tuple assignment
(Chapter 11): `a, b = b, a + b` gives `a` the old value of `b`, and `b` the sum of the old
values, because the right side is evaluated before any assignment happens. It also uses
`append` (Chapter 9) to add each number to the list.

````python
def fibonacci(n):
    result = []
    a, b = 0, 1
    while len(result) < n:
        result.append(a)
        a, b = b, a + b
    return result

print(fibonacci(0))
print(fibonacci(1))
print(fibonacci(4))
print(fibonacci(8))
````

The condition is `len(result) < n`: the loop repeats until the list holds `n` numbers. Each
iteration appends the current value of `a`, then moves `a` and `b` one step forward in the
sequence. For `fibonacci(0)`, the condition is false at the first check, so the function
returns an empty list.

No one knows how many guesses it will take, so this program needs a `while` loop. A `for` loop cannot do it: there is no list of guesses ready before the loop starts.

```python
import random


def guessing_game(low, high):
    num = random.randint(low, high)

        else:

        num_guesses = num_guesses + 1
    return num_guesses
```

## A loop that never ends

If the body never makes the condition false, the loop never ends. It is called an **infinite
loop**. Here is the factorial function with the update line left out:

```python
def factorial(n):
    result = 1
    while n > 1:
        result = result * n
    return result
```

`n` never changes, so `n > 1` stays true forever, and the loop repeats forever. The code
above is text, not a code cell, so it cannot run. If you run an infinite loop in Colab, stop
it with Runtime, then Interrupt execution.

Three mistakes cause most infinite loops: the body does not change a variable in the
condition, the body changes it in the wrong direction (for example, `n = n + 1` in a loop
that needs `n` to get smaller), or the condition itself is wrong (for example, `n < 1`
instead of `n > 1`). Question 4 asks you to find such a mistake.

<details>
<summary>Diving deeper: a for loop written as a while loop</summary>

Any `for` loop over a range can be written as a `while` loop. This `for` loop:

```python
for i in range(3):
    print(i)
```

displays the same values as this `while` loop:

```python
i = 0
while i < 3:
    print(i)
    i = i + 1
```

The `while` version needs three extra pieces: set `i` to `0` before the loop, check `i < 3`,
and add 1 to `i` in the body. The `for` loop does all three for you. For when to use each
kind of loop, see `When to use a while loop` above.

</details>

### Question 4 (medium): find the bug

<summary>Answer</summary>

```python
def is_prime(n):
    if n < 2:
        return False
    divisor = 2
    while divisor < n:
        if n % divisor == 0:
            return False
        divisor = divisor + 1
    return True
```

The loop tries each possible divisor from `2` up to `n - 1`. If one divides `n` evenly, `n` is not prime. If none does, `n` is prime. For `n` less than 2, the function returns `False` before the loop.
