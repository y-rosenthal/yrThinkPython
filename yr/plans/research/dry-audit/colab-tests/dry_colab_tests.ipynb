## T3. Recursion limit (E13)

````python
import sys

def depth(n):
    try:
        return depth(n + 1)
    except RecursionError:
        return n

print('T3 recursionlimit', sys.getrecursionlimit())
print('T3 depth reached', depth(0))
````
