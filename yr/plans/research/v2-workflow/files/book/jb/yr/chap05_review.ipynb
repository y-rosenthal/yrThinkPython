## Concepts covered

:::{admonition} Python Syntax and Semantics
:class: dropdown toggle-shown

**Integer division and modulus**

- `X // Y` divides and rounds down to an integer: `105 // 60` is `1`, and `-7 // 2` is `-4`.
- `X % Y` (the modulus operator) is the remainder when X is divided by Y: `105 % 60` is `45`.
- `X % Y == 0` is true when X is divisible by Y. `X % 10` is the last digit of X, and `X % 100` its last two digits.
- `(START + DURATION) % 12` is clock arithmetic: it wraps around after 12 (use `% 24` for a 24-hour clock).

**Boolean expressions**

- `True` and `False` are the two values of the type `bool`. They are not strings.
- The relational operators compare two values and produce `True` or `False`: `X == Y` (equal), `X != Y` (not equal), `X > Y`, `X < Y`, `X >= Y`, `X <= Y`.
- `=` assigns a value to a variable; `==` compares two values.
- `A and B` is true when both are true. `A or B` is true when either or both are true. `not A` is true when A is false. For example, `x > 0 and x < 10` is true when x is between 0 and 10.
- `LOW < X < HIGH` is shorthand for `LOW < X and X < HIGH`.
- Any nonzero number counts as true when used as a condition. (This can be confusing; it is better to use boolean expressions.)

**if statements**

- This runs the indented block only if the condition is true:
  ```
  if CONDITION:
      BLOCK
  ```
  For example, this displays `x is positive` if `x` is greater than `0`:
  ```python
  if x > 0:
      print('x is positive')
  ```
- A block must have at least one statement. `pass` is a statement that does nothing, used as a placeholder.
- With an `else` clause, exactly one of the two blocks runs:
  ```
  if CONDITION:
      BLOCK1
  else:
      BLOCK2
  ```
- A chained conditional checks the conditions in order and runs only the block of the first true one (or the `else` block, if none is true):
  ```
  if CONDITION1:
      BLOCK1
  elif CONDITION2:
      BLOCK2
  else:
      BLOCK3
  ```
  There can be any number of `elif` clauses. The `else` clause is optional, and must come last.
- Separate `if` statements, one after the other, are each checked: several of their blocks can run.
- An `if` statement can be nested inside a block of another one. A nested conditional can often be replaced by one condition with `and`.

**Recursion**

- A function can call itself. Each call has its own parameters and local variables.
- A recursive function needs a base case: a branch that does not make a recursive call. Each recursive call should move closer to the base case.
- Statements after a recursive call run after that call has finished, so they run in the opposite order of the calls.
- If the base case is never reached, the recursion goes on until Python's limit on the number of frames (about 3000) is exceeded, which is a `RecursionError`.

**Keyboard input**

- `input()` stops the program, waits for the user to type something and press Enter, and returns what they typed, as a string.
- `input(PROMPT)` first displays the prompt. `'\n'` in a string is a newline (a line break), so `input('Your name?\n')` lets the user type below the prompt.
- `int(input())` converts the input to an integer; if the user types something that is not an integer, it is a `ValueError`.

**Error messages**

- Indenting a line that should not be indented is an `IndentationError`.
- Leaving out the colon at the end of an `if` header is a `SyntaxError`, and so is using `=` instead of `==` in a condition.
- An error message says where the error was discovered, which may be after the line where the mistake is.

**The jupyturtle module** (used by one question)

- `make_turtle()` starts a drawing with the turtle in the middle, facing east; `forward(DISTANCE)` moves and draws (a negative distance moves backward); `left(DEGREES)` and `right(DEGREES)` turn; `penup()` and `pendown()` stop and restart drawing.

:::{admonition} Definitions, etc.
:class: dropdown toggle-shown

**Definitions**

- DEFINITION: The **modulus operator**, `%`, computes the remainder when one integer is divided by another.
- DEFINITION: A **boolean expression** is an expression whose value is either `True` or `False`.
- DEFINITION: A **relational operator** is one of the operators that compare their operands: `==`, `!=`, `>`, `<`, `>=` and `<=`.
- DEFINITION: A **logical operator** is one of the operators that combine boolean expressions: `and`, `or` and `not`.
- DEFINITION: A **conditional statement** is a statement that controls which statements run, depending on a condition.
- DEFINITION: A **condition** is the boolean expression in a conditional statement that decides which branch runs.
- DEFINITION: A **block** is one or more statements indented to show they are part of another statement.
- DEFINITION: A **branch** is one of the alternative blocks in a conditional statement.
- DEFINITION: A **chained conditional** is a conditional statement with a series of alternative branches (`if`, `elif`, ..., `else`).
- DEFINITION: A **nested conditional** is a conditional statement inside a branch of another conditional statement.
- DEFINITION: **Recursion** is calling the function that is currently running. A function that calls itself is **recursive**.
- DEFINITION: The **base case** is a branch of a recursive function that does not make a recursive call.
- DEFINITION: **Infinite recursion** is a recursion that has no base case or never reaches it. It ends with a runtime error.
- DEFINITION: A **newline** is a character, written `\n`, that makes a line break in a string.

**Other ideas**

- Nested conditionals are hard to read; avoid them when a simpler condition can do the same job.
- `TODO` in a comment is a conventional reminder of something still to be done.
- In a stack diagram of a recursive function, each call has its own frame with its own value of each parameter; the frame of the base case is at the bottom.
- When you see an infinite recursion, check that there is a base case and that every call is guaranteed to reach it.
- The most useful parts of an error message are the kind of error and where it happened. Read error messages carefully, but the actual mistake may be earlier in the code.
:::

```python
    if c > biggest:
```

### Question 14 (hard): infinite recursion

Run this cell to define `countdown_by_two`.

````python
def countdown_by_two(n):
    if n == 0:
        print('Blastoff!')
    else:
        countdown_by_two(n - 2)
````

Two correct ways to call it:

```python
countdown_by_two(4)
```
```text
Blastoff!
```

```python
countdown_by_two(10)
```
```text
Blastoff!
```

What happens with this call? How would you fix the function?

```python run
countdown_by_two(5)
```

:::{admonition} Answer
:class: dropdown

The calls get `n` = 5, 3, 1, -1, -3, ... and never `0`, so the base case is never reached.
This is infinite recursion. It ends with a `RecursionError` when there are too many frames on the stack:

```text
---------------------------------------------------------------------------
RecursionError                            Traceback (most recent call last)
/tmp/ipykernel_0/3608748751.py in <cell line: 0>()
----> 1 countdown_by_two(5)

/tmp/ipykernel_0/3568901665.py in countdown_by_two(n)
      3         print('Blastoff!')
      4     else:
----> 5         countdown_by_two(n - 2)

... last 1 frames repeated, from the frame below ...

/tmp/ipykernel_0/3568901665.py in countdown_by_two(n)
      3         print('Blastoff!')
      4     else:
----> 5         countdown_by_two(n - 2)

RecursionError: maximum recursion depth exceeded
```

Fix it by making the base case catch every number at or below zero:

```python
def countdown_by_two(n):
    if n <= 0:
        print('Blastoff!')
    else:
        countdown_by_two(n - 2)

countdown_by_two(5)
```
:::

### Question 16 (hard): write two recursive functions

Write two recursive functions that take a non-negative integer `n` and display its digits, one per line:
`digits_backward(n)` from the last digit to the first, and `digits_forward(n)` from the first to the last.

:::{admonition} Answer
:class: dropdown

A number with one digit (less than 10) is the base case. Otherwise, display the last digit and handle the rest recursively, or the other way around:

```python
def digits_backward(n):
    print(n % 10)
    if n >= 10:
        digits_backward(n // 10)

def digits_forward(n):
    if n >= 10:
        digits_forward(n // 10)
    print(n % 10)

digits_backward(472)

digits_forward(5)
```

The only difference is the order: `digits_forward` displays its digit *after* the recursive call has displayed all the earlier digits.

```python
def digits_forward(n):
    if n < 10:
        print(n)
    else:
        digits_forward(n // 10)
        print(n % 10)

```

`branch` draws a line, turns left 30 degrees and draws a smaller tree, turns right 60 degrees and draws another smaller tree,
then turns back and moves backward to where it started, facing the same way.
That last step is important: each call leaves the turtle where it found it, so the next branch starts in the right place.
The base case is a length of 10 or less, where nothing is drawn. The lengths are 70, 42, 25.2, 15.12, then 9.072, so there are four levels.
