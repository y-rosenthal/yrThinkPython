````python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://github.com/ramalho/jupyturtle/releases/download/2024-03/jupyturtle.py');

import math
from jupyturtle import make_turtle, forward, left
````

````python
# @title Helper for the Answers
def run_code(code):
    """Run code as a cell of its own; print an error's traceback as text instead of raising."""
    import sys
    from IPython import get_ipython
    ip = get_ipython()
    if ip is None:
        exec(code, globals())
        return
    def to_stderr(etype, evalue, stb):
        stb = getattr(stb, 'stb', stb)           # Colab passes a ColabTraceback for import errors
        print(ip.InteractiveTB.stb2text(stb), file=sys.stderr)
    ip._showtraceback = to_stderr
    try:
        ip.run_cell(code, store_history=False)
    finally:
        del ip._showtraceback
````

### Question 6 (medium): find the bug

`total` should be the sum of the numbers from 1 to `n`. What does the code display, and how do you fix it?

~~~python
def total(n):
    result = 0
    for i in range(n):
        result = result + i
    return result

print(total(4))
~~~

#### Answer

````python
# @title Output of Question 6
def total(n):
    result = 0
    for i in range(n):
        result = result + i
    return result

print(total(4))
````

`range(n)` stops before `n`. Count to `n + 1`:

<!-- derive: from=Q6 replace="range(n)" with="range(n + 1)" -->
```python
def total(n):
    result = 0
    for i in range(n + 1):
        result = result + i
    return result

print(total(4))
```
<!-- begin generated: output -->
```text

```
<!-- end generated: output -->

A fix that keeps its own input line:

```python

print(price * 2)
```

### Question 7 (hard): what happens?

~~~python

    else:
        countdown(n - 2)

countdown(3)
~~~

#### Answer

````python
# @title Output of Question 7
if 'run_code' in globals():
    run_code(r"""
def countdown(n):
    if n == 0:
        print('Blastoff!')
    else:
        countdown(n - 2)

countdown(3)
""")
else:
    print('Run the setup cells first: Runtime → Run all.')
````
