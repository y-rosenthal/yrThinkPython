## Concepts covered

<details open>
<summary><strong>Python Syntax and Semantics</strong></summary>

**Importing**

- `import MODULE_NAME` makes a module's functions available as `MODULE_NAME.FUNCTION_NAME(...)`. For example, after `import jupyturtle` you can call `jupyturtle.forward(100)`.
- `from MODULE_NAME import NAME1, NAME2` makes only the listed functions available, by their names alone. For example, after `from jupyturtle import forward` you can call `forward(100)`. It does not define the name `MODULE_NAME`.
- `import math` makes `math.pi` (the number pi, 3.14159...) available.

**The jupyturtle module** (turtle graphics in Jupyter and Colab)

- `make_turtle()` starts a new drawing, with the turtle in the middle, facing east (to the right).
- `make_turtle` takes optional keyword arguments: `delay=SECONDS` (pause after each step; default `0.2`; `0` is fastest), `width=PIXELS`, `height=PIXELS`, and `auto_render=False` (show the drawing only when `render()` is called).
- `forward(DISTANCE)` moves the turtle and draws a line. A negative distance moves it backward.
- `left(DEGREES)` and `right(DEGREES)` turn the turtle.
- `penup()` stops drawing as the turtle moves; `pendown()` starts drawing again.

**for loops and range**

- This repeats the indented code SOME_INTEGER times:
  ```
  for i in range(SOME_INTEGER):
      SOME_CODE
      SOME_MORE_CODE
  ```
  For example, this prints `hello` three times:
  ```python
  for i in range(3):
      print('hello')
  ```
- `range(SOME_INTEGER)` is shorthand for the integers `0, 1, 2, ...` up to SOME_INTEGER minus 1. The loop variable (`i`) takes these values in turn: in `for i in range(4):`, `i` is `0`, then `1`, then `2`, then `3`.
- `range` needs an integer: `range(4.5)` is a `TypeError`. `range(0)` and `range(-4)` produce no values, so the loop body runs zero times.

**Functions with parameters**

- This defines a function with two parameters:
  ```
  def FUNCTION_NAME(PARAMETER1, PARAMETER2):
      BODY
  ```
  The body runs only when the function is called.
- `FUNCTION_NAME(ARGUMENT1, ARGUMENT2)` calls it: the arguments are assigned to the parameters, in order.
- A function can call other functions. Each call finishes before the next statement runs.

**Keyword arguments**

- `FUNCTION_NAME(PARAMETER_NAME=VALUE)` passes an argument by naming its parameter, as in `polygon(n=7, length=30)`. With keyword arguments, the order does not matter.
- Positional arguments (without names) must come first: `polygon(n=7, 30)` is a `SyntaxError`.
- Each parameter gets one value: `polygon(30, n=7)` is a `TypeError`. So is a name that is not a parameter: `polygon(7, size=30)`.

**Local variables**

- Parameters, and variables assigned inside a function, are local to one call of the function. The same name in two functions means two different variables.
- Assigning a new value to a parameter does not change the caller's variable.

**Division**

- `/` always produces a float: `360 / 4` is `90.0`.
- Dividing by zero, as in `360 / 0`, is a `ZeroDivisionError`.

**Docstrings and multiline strings**

- `"""TEXT"""` (triple quotes) is a string that can span several lines.
- A string on the first line of a function's body is the function's docstring:
  ```
  def FUNCTION_NAME(PARAMETERS):
      """WHAT THE FUNCTION DOES"""
      BODY
  ```
- A docstring is not displayed when the function runs. `help(FUNCTION_NAME)` displays it.

</details>

<details open>
<summary><strong>Definitions, etc.</strong></summary>

**Definitions**

- DEFINITION: **Encapsulation** is wrapping a sequence of statements in a function definition.
- DEFINITION: **Generalization** is replacing something unnecessarily specific (such as a number) with something more general (such as a parameter).
- DEFINITION: A **keyword argument** is an argument that includes the name of the parameter.
- DEFINITION: **Refactoring** is changing working code to improve it (for example, its function interfaces) without changing what it does.
- DEFINITION: The **interface** of a function is how it is used: its name, its parameters, and what it does.
- DEFINITION: The **implementation** of a function is how it does what it does.
- DEFINITION: **Interface design** is deciding on a function's interface, including which parameters it takes.
- DEFINITION: A **docstring** is a string at the beginning of a function that explains its interface.
- DEFINITION: A **multiline string** is a string in triple quotes, which can span more than one line.
- DEFINITION: A **precondition** is a requirement the caller must satisfy before a function starts.
- DEFINITION: A **postcondition** is something a function guarantees when it ends: its intended effect and any side effects.
- DEFINITION: A **stack diagram** shows the function calls in progress, one frame per call, with each call's parameters and their values.
- DEFINITION: A **development plan** is a process for writing programs.

**Other ideas**

- Encapsulation gives code a name that documents it, and calling a function twice is shorter than copying its code.
- Two functions can have the same interface and different implementations.
- The "encapsulation and generalization" development plan:
  1. Write a small program with no function definitions.
  2. When it works, put a coherent piece of it in a function and give the function a name.
  3. Generalize the function by adding parameters.
  4. Repeat steps 1 to 3 until you have a set of working functions.
  5. Look for chances to refactor, for example by replacing similar code in several places with one general function.
- That plan is useful when you don't know in advance how to divide a program into functions.
- Refactoring is normal: often you don't know enough at the start of a project to design all the functions.
- A good docstring says what the function does (not how), what each parameter means, and each parameter's type if it is not obvious. If a function is hard to explain, its interface may need improving.
- An interface is like a contract: the caller provides certain arguments and the function does certain work.
- If the caller breaks a precondition, the bug is in the caller. If the preconditions hold but a postcondition does not, the bug is in the function.

</details>

### Question 19 (hard): generalize a function

This function draws an arc (part of a circle) with the given `radius`, spanning `angle` degrees, as 30 short straight lines:

def arc(radius, angle):
    arc_length = 2 * math.pi * radius * angle / 360
    n = 30
    length = arc_length / n
    step_angle = angle / n
    for i in range(n):

```python
import math
from jupyturtle import make_turtle, forward, left

def arc_n(radius, angle, n):
    arc_length = 2 * math.pi * radius * angle / 360
    length = arc_length / n
    step_angle = angle / n
    for i in range(n):
        forward(length)
        left(step_angle)

def circle(radius):
    circumference = 2 * math.pi * radius
    n = int(circumference / 3) + 10
    arc_n(radius, 360, n)

make_turtle(delay=0)
circle(30)
```

For `radius` 30 the circumference is about 188, so `n` is 62 + 10 = 72 lines; for `radius` 300 it would be 628 + 10 = 638.
Adding `10` guarantees at least 10 lines, even for a tiny circle.

Another way to say "at least 10" is `max(10, int(circumference / 3))`, if you know the `max` function.
