## Concepts covered

<details open>
<summary><strong>Python Syntax and Semantics</strong></summary>

**Return values**

- This defines a function that returns a value:
  ```
  def FUNCTION_NAME(PARAMETERS):
      BODY
      return EXPRESSION
  ```
  A call to the function is an expression whose value is the return value. For example, after
  ```python
  def double(x):
      return x * 2
  ```
  `double(5)` is `10`, `y = double(5)` assigns `10` to `y`, and `double(5) + 1` is `11`.
- A `return` statement ends the function immediately; any statements after it in the same block never run.
- A function returns `None` if it ends without running a `return` statement (or runs `return None`). `None` is a special value, like `True` and `False`.
- `print` displays a value; `return` hands a value back to the caller. A function that only prints returns `None`, so its result cannot be used in an expression.
- In a notebook, a cell whose last expression is `None` displays nothing; `print(None)` displays `None`.
- Calling a function without using its return value (not assigning it, printing it or using it in an expression) throws the value away.
- Local variables of a function, including the variable it returns, cannot be used outside the function: that is a `NameError`.
- A function can have several `return` statements, for example one in each branch of an `if` statement. Every possible path through the function should reach a `return`; a path that does not returns `None`.

**Boolean functions**

- A function can return `True` or `False`. A comparison already has a boolean value, so it can be returned directly: `return x % y == 0` instead of an `if` statement that returns `True` or `False`.
- A boolean function can be used directly as a condition: `if is_divisible(6, 2):`. Comparing its result with `== True` is unnecessary.

**Recursion with return values**

- A recursive function can return a value: the base case returns a value directly, and the other cases combine the result of a recursive call into their return value. For example:
  ```python
  def factorial(n):
      if n == 0:
          return 1
      else:
          return n * factorial(n - 1)
  ```
- If an argument can skip past the base case (for example, `factorial(1.5)` calls `factorial(0.5)`, then `factorial(-0.5)`, ...), the result is infinite recursion and a `RecursionError`.

**Checking types**

- `isinstance(VALUE, TYPE)` returns `True` if the value has that type: `isinstance(3, int)` is `True` and `isinstance(1.5, int)` is `False`.
- A function can check its arguments first, and display an error message and return `None` if they are not valid.

**The math module**

- `import math` makes `math.sqrt(X)` (square root, returns a float) and `math.pi` available.

</details>

<details open>
<summary><strong>Definitions, etc.</strong></summary>

**Definitions**

- DEFINITION: A **return value** is the result of a function. If a function call is used as an expression, the return value is the value of the expression.
- DEFINITION: A **pure function** is a function that does not display anything or have any other effect, other than returning a value.
- DEFINITION: **Dead code** is part of a program that can never run, often because it comes after a `return` statement.
- DEFINITION: **Incremental development** is a way of writing programs by adding and testing only a small amount of code at a time.
- DEFINITION: **Scaffolding** is code used while developing a program (such as `print` calls that show intermediate values) that is not part of the final version.
- DEFINITION: A language, or part of a language, is **Turing complete** if it can perform any computation that can be described by an algorithm. Python with functions, conditionals, return values and recursion is Turing complete.
- DEFINITION: **Input validation** is checking the parameters of a function to make sure they have the correct types and values.

**Other ideas**

- Incremental development:
  1. Start with a working program (for example, a function that always returns `0`), make small changes, and test after every change.
  2. Use variables to hold intermediate values so you can display and check them.
  3. Once the program works, remove the scaffolding.
- When testing a function, choose arguments for which you know the right answer (for example, points 3 and 4 apart, whose distance is 5).
- The "leap of faith": when reading a function call, including a recursive call, assume the function works correctly and returns the right result, instead of following the flow of execution into it.
- A recursive definition (such as: the factorial of 0 is 1, and the factorial of n is n times the factorial of n minus 1) can be translated directly into a recursive function.
- If a function is not working, there are three possibilities: its arguments are wrong (a precondition is violated), the function is wrong (a postcondition is violated), or the caller uses the return value wrongly (or not at all).
- `print` calls at the beginning and end of a function (showing the parameters and the return value) make the flow of execution visible.

</details>

### Question 9 (medium): what is displayed?

```python
def half(n):
    if not isinstance(n, int):
        print('half needs an integer')
        return None
    return n // 2

print(half(9))
print(half(9.0))
```

#### Answer

```text
4
half needs an integer
None
```

`isinstance(9, int)` is `True`, so `half(9)` returns `9 // 2`, which is `4`.
`9.0` is a float, not an integer, so `half(9.0)` displays the message and returns `None`, which the outer `print` displays.
Checking the arguments like this is called input validation.

````python
def half(n):
    if not isinstance(n, int):
        print('half needs an integer')
        return None
    return n // 2

print(half(9))
print(half(9.0))
````

def larger(a, b):

### Question 13 (hard): what is displayed?

```python
def power(b, e):
    print('power', b, e)
    if e == 0:
        return 1
    result = b * power(b, e - 1)
    print('returning', result)
    return result

print(power(2, 3))
```

#### Answer

```text
power 2 3
power 2 2
power 2 1
power 2 0
returning 2
returning 4
returning 8
8
```

Each call displays its arguments and then waits for its recursive call to return. The call with `e == 0` is the base case: it returns `1` without displaying `returning`.
Then the calls finish in the opposite order: `2 * 1` is `2`, `2 * 2` is `4`, `2 * 4` is `8`.
The last line is displayed by the `print` outside the function. The `print` calls inside `power` are scaffolding, useful to see the flow of execution.

````python
def power(b, e):
    print('power', b, e)
    if e == 0:
        return 1
    result = b * power(b, e - 1)
    print('returning', result)
    return result

print(power(2, 3))
````

### Question 14 (hard): what is displayed?

Predict what this code displays. What does `mystery` compute, in general?

```python
def mystery(n):
    if n < 10:
        return n
    return n % 10 + mystery(n // 10)

print(mystery(7), mystery(123), mystery(9999))
```

#### Answer

```text
7 6 36
```

`mystery` returns the sum of the digits of `n`: `n % 10` is the last digit and `n // 10` is the number without its last digit.
With the leap of faith: assuming `mystery(12)` correctly returns `1 + 2`, `mystery(123)` returns `3 + mystery(12)`, which is `6`.
A number with one digit (less than 10) is the base case.

````python
def mystery(n):
    if n < 10:
        return n
    return n % 10 + mystery(n // 10)

print(mystery(7), mystery(123), mystery(9999))
````
