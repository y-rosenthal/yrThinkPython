## Concepts covered

<details open>
<summary><strong>Python Syntax and Semantics</strong></summary>

**Creating dictionaries**

- `{}` is an empty dictionary: `len({})` is <!--=-->`0`.
- This creates a dictionary with three items:
  ```
  {'zero': 0, 'one': 1, 'two': 2}
  ```
  Each item is `KEY: VALUE`, and items are separated by commas. The keys can be almost any type, and the values can be any type.
- `dict()` makes an empty dictionary. `dict(DICT)` makes a copy of `DICT`: `dict({'a': 1})` is <!--=-->`{'a': 1}`.

**Adding and changing items**

- This adds a new item to the dictionary:
  ```
  DICT[KEY] = VALUE
  ```
  For example, after this code `d` is <!--= d = {}; d['a'] = 1; d -->`{'a': 1}`:
  ```python
  d = {}
  d['a'] = 1
  ```
- Assigning to a key that is already there replaces its value; it does not add a new item. After this code `d` is <!--= d = {'a': 1}; d['a'] = 99; d -->`{'a': 99}`:
  ```python
  d = {'a': 1}
  d['a'] = 99
  ```

**Looking up a value**

- `DICT[KEY]` returns the value for `KEY`: `{'a': 1, 'b': 2}['b']` is <!--=-->`2`.
- Looking up a key that is not in the dictionary is a `KeyError`.
- `len(DICT)` returns the number of items: `len({'a': 1, 'b': 2})` is <!--=-->`2`.

**The `in` operator**

- `KEY in DICT` is `True` if `KEY` is a key in the dictionary: `'a' in {'a': 1}` is <!--=-->`True`.
- `in` checks keys only, not values: `1 in {'a': 1}` is <!--=-->`False`.
- `DICT.values()` gives the values, so `VALUE in DICT.values()` checks whether a value is in the dictionary: `1 in {'a': 1}.values()` is <!--=-->`True`.

**The `get` method**

- `DICT.get(KEY, DEFAULT)` returns the value for `KEY` if it is there, and `DEFAULT` if it is not: `{'a': 1}.get('b', 0)` is <!--=-->`0`.

**Looping through a dictionary**

- This runs the body once for each key in the dictionary, in the order the keys were added:
  ```
  for KEY in DICT:
      BODY
  ```
- This runs the body once for each value:
  ```
  for VALUE in DICT.values():
      BODY
  ```
- To get the keys and the values, loop through the keys and look up each value with `DICT[KEY]`.

**Lists and dictionaries**

- A list can be a value: `{'a': [1, 2]}` has a list as its value.
- A list cannot be a key, because a list is mutable and keys must be hashable. Using a list as a key is a `TypeError`.
- A dictionary is mutable, so it can be a value but not a key.
- A value can itself be a dictionary: `{'ann': {'age': 20}}` is a dictionary with a dictionary as its value. To look up a value in the inner dictionary, use two brackets: `{'ann': {'age': 20}}['ann']['age']` is <!--=-->`20`.

**The counter pattern**

- To count how many times each letter appears, use a dictionary that maps each letter to its count. The first time a letter appears, add it with a count of 1; after that, add 1 to its count:
  ```python
  if letter not in counter:
      counter[letter] = 1
  else:
      counter[letter] += 1
  ```
- `COUNTER[KEY] += 1` adds 1 to the value for `KEY`.

**A hash table**

- A dictionary is a hash table, so looking up a key takes about the same time no matter how many items the dictionary has. `in` on a list takes time proportional to the length of the list, so for a long list it is much slower.

**Accumulating and filtering**

- The accumulator pattern: start with an empty list (or 0) and add to it each time through the loop.
- Filtering: loop through a sequence and select (or omit) some of its elements.

**Memos**

- A memo is a computed value stored in a dictionary, so it does not have to be computed again. Before computing, check `if N in KNOWN`.

<summary><strong>Definitions, etc.</strong></summary>

**Definitions**

- DEFINITION: A **dictionary** is an object that contains key-value pairs, also called items.
- DEFINITION: An **item** is, in a dictionary, another name for a key-value pair.
- DEFINITION: A **key** is an object that appears in a dictionary as the first part of a key-value pair.
- DEFINITION: A **value** is an object that appears in a dictionary as the second part of a key-value pair. This is more specific than our previous use of the word "value".
- DEFINITION: A **mapping** is a relationship in which each element of one set corresponds to an element of another set.
- DEFINITION: A **hash table** is a collection of key-value pairs organized so that we can look up a key and find its value efficiently.
- DEFINITION: A **hashable** object is one that can be a key in a dictionary. Immutable types like integers, floats and strings are hashable. Mutable types like lists and dictionaries are not.
- DEFINITION: A **hash function** is a function that takes an object and computes an integer that is used to locate a key in a hash table.
- DEFINITION: An **accumulator** is a variable used in a loop to add up or accumulate a result.
- DEFINITION: **Filtering** is looping through a sequence and selecting or omitting elements.
- DEFINITION: A **call graph** is a diagram that shows every frame created during the execution of a program, with an arrow from each caller to each callee.
- DEFINITION: A **memo** is a computed value stored to avoid unnecessary future computation.

**Other ideas**

- Do not use `dict` as a variable name, because it is the name of a built-in function.
- When you work with a dictionary, remember that `in` checks keys, not values.

<!-- begin generated: define -->
Run this cell to define `fibonacci_memo` and assign `known`.
<!-- end generated: define -->

````python
known = {0: 0, 1: 1}

def fibonacci_memo(n):
    if n in known:
        return known[n]
    res = fibonacci_memo(n - 1) + fibonacci_memo(n - 2)
    known[n] = res
    return res
````
