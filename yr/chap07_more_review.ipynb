### Question 1 (easy): write a function

Write a function called `count_evens` that takes `n` and returns how many even numbers are in `range(n)`, starting at 0.

For example, `count_evens(10)` is `5` (0, 2, 4, 6, 8). `count_evens(1)` is `1` (just 0 — is 0 even? Yes, `0 % 2` is `0`).

Start from this header (replace `...` with the body):

```python
def count_evens(n):
    ...
```

````python
# Your code here
````

<details>
<summary>Answer</summary>

```python
def count_evens(n):
    count = 0
    for i in range(n):
        if i % 2 == 0:
            count += 1
    return count
```

### Question 3 (medium): write a function

Write a function called `fizzbuzz` that prints, one per line, for 1 through `n`: `'FizzBuzz'` for a number divisible by both 3 and 5, `'Fizz'` for a number divisible by 3 only, `'Buzz'` for a number divisible by 5 only, and the number itself otherwise.

For example, `fizzbuzz(15)` ends with `..., 13, 14, FizzBuzz` (each on its own line).

Question to think about first: what goes wrong if you test divisibility by 3 before you test divisibility by both?

Start from this header (replace `...` with the body):

```python

    ...
```

````python
# Your code here
````

<details>
<summary>Answer</summary>

```python
def fizzbuzz(n):
    for i in range(1, n+1):
        if i % 3 == 0 and i % 5 == 0:
            print('FizzBuzz')
        elif i % 3 == 0:
            print('Fizz')
        elif i % 5 == 0:
            print('Buzz')
        else:
            print(i)
```

Test both first, then 3, then 5, then `else`. This version prints instead of building a list, because lists are not taught until Chapter 9.

For example, `describe_number(4)` is `'positive even'`; `describe_number(-3)` is `'negative odd'`; `describe_number(0)` is `'zero'`.

<details>
<summary>Answer</summary>

```python
def describe_number(n):
    if n > 0:
        if n % 2 == 0:
            return 'positive even'
        else:
            return 'positive odd'
    elif n < 0:
        if n % 2 == 0:
            return 'negative even'
        else:
            return 'negative odd'
    else:
        return 'zero'
```

An outer `if`/`elif`/`else` on the sign, with an inner `if`/`else` on evenness inside the positive and negative branches. Zero must be handled separately from evenness — otherwise 0 would be called 'positive even'.

### Question 7 (medium): write a function

Write a function called `classify_letters` that takes a string `s` and returns three numbers: how many vowels, how many consonants, and how many other characters are in `s`.

For example, `classify_letters('Hi! 123')` is `(1, 1, 5)`.

```python
def classify_letters(s):
    ...
```

````python
# Your code here
````

<details>
<summary>Answer</summary>

```python
def classify_letters(s):
    vowels = consonants = other = 0
    for ch in s:
        if ch in 'aeiouAEIOU':
            vowels += 1
        elif ('a' <= ch <= 'z') or ('A' <= ch <= 'Z'):
            consonants += 1
        else:
            other += 1
    return vowels, consonants, other
```

Order matters: vowels must be tested before consonants, or every vowel would be counted as a consonant.

</details>

### Question 8 (hard): write a function

Write a function called `first_even_over` that takes `n` and `limit` and returns the first even number from `n` through `limit`, or `None` if there is none.

For example, `first_even_over(7, 20)` is `8`; `first_even_over(7, 7)` is `None`.

Start from this header (replace `...` with the body):

```python
def first_even_over(n, limit):
    ...
```

````python
# Your code here
````

<details>
<summary>Answer</summary>

```python
def first_even_over(n, limit):
    for i in range(n, limit+1):
        if i % 2 == 0:
            return i
    return None
```

The `return` inside the loop is the search pattern from Chapter 7: the first value that passes the test ends the search.
