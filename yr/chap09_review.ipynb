## Concepts covered

<summary><strong>Python Syntax and Semantics</strong></summary>

**Creating lists**

- This creates a list of the values in the brackets, in order:
  ```
  [VALUE1, VALUE2, VALUE3]
  ```
  For example, this makes `cheeses` refer to a list of three strings:
  ```python
  cheeses = ['Cheddar', 'Edam', 'Gouda']
  ```
- The elements of a list can have any type, and different types in one list. A list can be an element of another list: `['spam', 2.0, 5, [10, 20]]` has a string, a float, an integer and a list.
- `[]` is an empty list.
- `len(LIST)` returns the number of elements in the list. A nested list counts as one element: `len(['spam', 2.0, 5, [10, 20]])` is <!--=-->`4`, and `len([])` is <!--=-->`0`.
- `list(SEQUENCE)` returns a new list with the elements of the sequence. The elements of a string are its characters: `list('spam')` is <!--=-->`['s', 'p', 'a', 'm']`.

**Indexes and slices**

- `LIST[INDEX]` is the element at that index. The first element has index `0`, and a negative index counts back from the end: `['a', 'b', 'c'][0]` is <!--=-->`'a'`, and `['a', 'b', 'c'][-1]` is <!--=-->`'c'`. Any integer expression can be an index.
- Reading or writing an element that does not exist, such as `['a', 'b', 'c'][3]`, is an `IndexError`.
- Lists are mutable. This replaces the element at INDEX with VALUE:
  ```
  LIST[INDEX] = VALUE
  ```
  For example, this changes the second element of `numbers` from `123` to `17`:
  ```python
  numbers = [42, 123]
  numbers[1] = 17
  ```
- `LIST[START:END]` returns a new list with the elements from index START up to END minus 1: `['a', 'b', 'c', 'd'][1:3]` is <!--=-->`['b', 'c']`. Without START the slice starts at the beginning, and without END it goes to the end: `['a', 'b', 'c', 'd'][2:]` is <!--=-->`['c', 'd']`.
- `LIST[:]` and `list(LIST)` return a copy of the list: a new list with the same elements.
- `VALUE in LIST` is `True` if the value is an element of the list. It does not look inside nested lists: `10 in ['spam', [10, 20]]` is <!--=-->`False`.

**Operators and built-in functions**

- `LIST1 + LIST2` returns a new list with the elements of both lists: `[1, 2] + [3, 4]` is <!--=-->`[1, 2, 3, 4]`. Neither list changes.
- `LIST * SOME_INTEGER` returns a new list with the elements repeated: `['spam'] * 3` is <!--=-->`['spam', 'spam', 'spam']`.
- No other arithmetic operators work with lists.
- `sum(LIST)` adds up the elements, and `min(LIST)` and `max(LIST)` return the smallest and the largest element: `sum([3, 1, 4])` is <!--=-->`8`, and `max([3, 1, 4])` is <!--=-->`4`.

**List methods**

- `LIST.append(VALUE)` adds the value to the end of the list.
- `LIST.extend(OTHER_LIST)` adds all the elements of OTHER_LIST to the end of the list.
- `LIST.pop(INDEX)` removes the element at that index, and returns it.
- `LIST.remove(VALUE)` removes the first element that is equal to the value. If the value is not in the list, it is a `ValueError`.
- These methods change the list. `append`, `extend` and `remove` return `None`. String methods are the opposite: they return a new string and do not change the original.
- So `t = t.remove(3)` is a mistake: it makes `t` refer to `None`. After it, `t.remove(2)` is an `AttributeError`, because `None` has no attribute `remove`.

**Lists and strings**

- `STRING.split()` returns a list of the words in the string (the parts between whitespace: spaces, tabs and newlines): `'pining for the fjords'.split()` is <!--=-->`['pining', 'for', 'the', 'fjords']`.
- `STRING.split(DELIMITER)` splits the string at each DELIMITER instead: `'ex-parrot'.split('-')` is <!--=-->`['ex', 'parrot']`.
- `DELIMITER.join(LIST_OF_STRINGS)` returns one string with the elements of the list, with DELIMITER between them: `' '.join(['pining', 'for', 'the', 'fjords'])` is <!--=-->`'pining for the fjords'`. With the empty string as the delimiter, nothing goes between them: `''.join(['a', 'b', 'c'])` is <!--=-->`'abc'`.
- `STRING.strip(CHARACTERS)` returns a new string without those characters at its beginning and end: `'plumage!'.strip('!')` is <!--=-->`'plumage'`.
- `open(FILE_NAME).read()` returns the whole file as one string. `open(FILE_NAME).read().split()` is a list of the words in the file.

**Looping through a list**

- This runs the body once for each element of the list, in order, with the loop variable referring to that element:
  ```
  for LOOP_VARIABLE in LIST:
      BODY
  ```
  For example, this displays `Cheddar`, `Edam` and `Gouda` on separate lines:
  ```python
  for cheese in ['Cheddar', 'Edam', 'Gouda']:
      print(cheese)
  ```
- A loop through an empty list never runs the body.
- To make a list in a loop, start with an empty list and append to it:
  ```
  NEW_LIST = []
  for LOOP_VARIABLE in SEQUENCE:
      NEW_LIST.append(VALUE)
  ```

**Sorting and reversing**

- `sorted(SEQUENCE)` returns a new list with the elements in increasing order. The original does not change. It works with any sequence, also a string: `sorted('cab')` is <!--=-->`['a', 'b', 'c']`, and `''.join(sorted('cab'))` is <!--=-->`'abc'`.
- `reversed(SEQUENCE)` returns the elements in reverse order, as a `reversed` object. Use `list` or `join` to get a list or a string: `list(reversed([1, 2, 3]))` is <!--=-->`[3, 2, 1]`, and `''.join(reversed('parrot'))` is <!--=-->`'torrap'`.

**Objects, equality and aliasing**

- `A == B` is `True` if A and B have the same value. `A is B` is `True` if A and B refer to the same object.
- Each list display (`[...]`) makes a new list object: `[1, 2, 3] == [1, 2, 3]` is <!--=-->`True`, but `[1, 2, 3] is [1, 2, 3]` is <!--=-->`False`.
- `b = a` does not copy the object: it makes `b` refer to the same object as `a`. If the object is a list, a change made with one name is seen with the other name. For example, after this code `a` is <!--= a = [1, 2, 3]; b = a; b[0] = 5; a -->`[5, 2, 3]`:
  ```python
  a = [1, 2, 3]
  b = a
  b[0] = 5
  ```
- When a list is passed to a function, the parameter refers to the same list. If the function changes the list, the caller sees the change after the function returns.

</details>

<details open>
<summary><strong>Definitions, etc.</strong></summary>

**Definitions**

- DEFINITION: A **list** is an object that contains a sequence of values.
- DEFINITION: An **element** is one of the values in a list or another sequence.
- DEFINITION: A **nested list** is a list that is an element of another list.
- DEFINITION: An **empty list** is a list with no elements, `[]`.
- DEFINITION: A **mutable** object can be changed; an **immutable** object, such as a string, cannot. Lists are mutable.
- DEFINITION: A **delimiter** is a character or string that shows where a string is split, or what goes between the strings that `join` joins.
- DEFINITION: Two objects are **equivalent** if they have the same value (`==`).
- DEFINITION: Two objects are **identical** if they are the same object (`is`). Identical objects are always equivalent; equivalent objects are not always identical.
- DEFINITION: A **reference** is the association between a variable and its value (an object).
- DEFINITION: An object is **aliased** if more than one variable refers to it.
- DEFINITION: An **attribute** is one of the named values (variables or methods) associated with an object.

**Other ideas**

- Do not use `list` as a variable name, because it is the name of a built-in function.
- Read a file once and keep its words in a list, instead of reading the file each time you search it.
- Aliasing is useful, but it causes errors that are difficult to find. When you work with mutable objects, it is safer to avoid aliasing (for example, make a copy with `[:]`). For immutable objects such as strings, aliasing is almost never a problem.
- Passing a list to a function makes an alias: the parameter and the caller's variable refer to the same list.
- Most list methods change the list and return `None`. Do not assign their result to the list variable.
- If you see an `AttributeError` that says `'NoneType' object has no attribute`, look back through the program for a list method whose result was assigned.

</details>

#### Answer

Use `split` to get a list of the words, then take the first letter of each word:

```python
def acronym(phrase):
    result = ''
    for word in phrase.split():

    return result
```

```python
    return ''.join(letters).upper()
```

Then try: `print(running_totals([1, 2, 3, 4]))`.
<!-- end generated: then-try -->

Another way: for each index `i`, the element is the sum of the slice from the beginning up to index `i`:

```python
def running_totals(numbers):
    result = []
    for i in range(len(numbers)):
        result.append(sum(numbers[:i + 1]))
    return result
```

The second way adds up the numbers again for each element, so it is slower for long lists.

#### Answer

````python
# @title Output of Question 16
def add_zero(lst):
    lst.append(0)

def add_one(lst):
    lst = lst + [1]

numbers = [5]
add_zero(numbers)
add_one(numbers)
print(numbers)
````

In both functions, the parameter `lst` first refers to the same list as `numbers`.

~~~python
row = [0, 0]

grid[0][1] = 5

copies[0][0] = 7
~~~

#### Answer

````python
# @title Output of Question 17
row = [0, 0]
grid = [row, row]
grid[0][1] = 5
print(grid)
print(row)
copies = [row[:], row[:]]
copies[0][0] = 7
print(copies)
````

Both elements of `grid` refer to the same list, `row`: the list is aliased three times.
So `grid[0][1] = 5` changes `row`, and the change is seen in both elements of `grid`.

#### Answer

`remove` removes only the first element that is equal to its argument, so call it once for each element that is equal to the value.
Loop through a copy of the list (`t[:]`), because the loop must not skip elements when `remove` changes `t`:

```python
def remove_all(t, value):
    for x in t[:]:
        if x == value:
            t.remove(x)
```
<!-- begin generated: then-try -->
Then try: `numbers = [1, 1, 2]`, `remove_all(numbers, 1)` and `print(numbers)`.
<!-- end generated: then-try -->

Another way: the list has at most `len(t)` elements equal to the value, so call `remove` up to `len(t)` times, as long as the value is in the list:

```python
def remove_all(t, value):
    for i in range(len(t)):
        if value in t:
            t.remove(value)
```

A loop through `t` itself skips elements: after `remove`, each later element moves back one position, so the loop never sees the element just after the one it removed.

### Question 20 (hard): write a function

Write a function called `has_duplicates` that takes a list and returns `True` if any element appears in it more than once, and `False` otherwise.

<!-- begin generated: header -->
Start from this header (replace `...` with the body):

```python
def has_duplicates(t):
    ...
```
<!-- end generated: header -->

**Examples**

```python
print(has_duplicates([1, 2, 3, 2]), has_duplicates(['a', 'b']))
```
<!-- begin generated: example -->
```text
True False
```
<!-- end generated: example -->

```python
print(has_duplicates([]), has_duplicates(list('banana')))
```
<!-- begin generated: example -->
```text
False True
```
<!-- end generated: example -->

````python
# Your code here
````

#### Answer

For each index `i`, check whether the element at `i` is also in the rest of the list, after `i`:

```python
def has_duplicates(t):
    for i in range(len(t)):
        if t[i] in t[i + 1:]:
            return True
    return False
```
<!-- begin generated: then-try -->
Then try: `print(has_duplicates([1, 2, 3, 2]), has_duplicates(['a', 'b']))`.
<!-- end generated: then-try -->

Another way: after sorting, equal elements are next to each other, so compare each element with the next one:

```python
def has_duplicates(t):
    s = sorted(t)
    for i in range(len(s) - 1):
        if s[i] == s[i + 1]:
            return True
    return False
```

Both are linear searches: they return `True` as soon as they find a duplicate, and `False` only after the loop. For an empty list, `range(len(s) - 1)` is `range(-1)`, which produces no values.
