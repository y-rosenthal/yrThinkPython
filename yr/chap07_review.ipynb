## Concepts covered

<details open>
<summary><strong>Python Syntax and Semantics</strong></summary>

**Looping through a string**

- This runs the body once for each character of the string, in order, with the loop variable referring to that character:
  ```
  for LOOP_VARIABLE in STRING:
      BODY
  ```
  For example, this displays `G`, `a`, `d` on separate lines:
  ```python
  for letter in 'Gad':
      print(letter)
  ```
- `print(VALUE, end=' ')` displays the value followed by a space instead of a newline, so the next `print` continues on the same line.

**Updating variables**

- A new assignment makes an existing variable refer to a new value. `x = x + 1` is an update: it computes the right side with the current value of `x`, then assigns the result to `x`.
- Updating a variable that does not exist yet, such as `z = z + 1`, is a `NameError`. Initialize it first, for example with `z = 0`.
- Augmented assignment: `x += N` is the same as `x = x + N`. Likewise `x -= N` and `x *= N`.
- A counter is a variable initialized to `0` before a loop and incremented inside it:
  ```
  count = 0
  for LOOP_VARIABLE in SEQUENCE:
      if CONDITION:
          count += 1
  ```

**The in operator and string methods**

- `STRING1 in STRING2` is `True` if STRING1 appears in STRING2: `'e' in 'Gadsby'` is <!--=-->`False`, and `'ads' in 'Gadsby'` is <!--=-->`True`. Uppercase and lowercase letters are different: `'g' in 'Gadsby'` is <!--=-->`False`.
- A method is a function associated with a value, called with a dot: `VALUE.METHOD_NAME(ARGUMENTS)`.
- `STRING.lower()` returns a new string with the letters in lowercase: `'Gadsby'.lower()` is `'gadsby'`. It does not change the original string; to keep the result, assign it, as in `word = word.lower()`.
- `STRING.strip()` returns a new string without the whitespace (spaces, tabs and newlines) at its beginning and end: `'  hi \n'.strip()` is `'hi'`.

**Reading files**

- `open(FILE_NAME)` returns a file object for reading the file.
- `FILE_OBJECT.readline()` returns the next line of the file as a string, including the newline at its end (`'\n'`). The file object keeps track of where it is, so each call returns the next line.
- This runs the body once for each line of the file:
  ```
  for line in open(FILE_NAME):
      word = line.strip()
      BODY
  ```

**Linear search**

- A linear search loops through a sequence and returns as soon as it finds what it is looking for; if the loop ends without finding it, the function returns after the loop:
  ```python
  def has_e(word):
      for letter in word:
          if letter == 'e':
              return True
      return False
  ```
- The `return False` must be after the loop, not in an `else` inside it: otherwise the function returns after checking only the first letter.

**Doctests**

- A docstring can contain tests: a line starting with `>>> ` followed by an expression, and on the next line the value the expression should have:
  ```
  def FUNCTION_NAME(PARAMETERS):
      """WHAT THE FUNCTION DOES

      >>> FUNCTION_NAME(ARGUMENTS)
      EXPECTED_VALUE
      """
      BODY
  ```
- `run_docstring_examples(FUNCTION, globals(), name=FUNCTION.__name__)`, after `from doctest import run_docstring_examples`, runs a function's doctests. It displays nothing if all of them pass; for each test that fails, it displays the example, the expected value and the value it got.

</details>

<summary><strong>Definitions, etc.</strong></summary>

**Definitions**

- DEFINITION: A **loop variable** is the variable defined in the header of a `for` loop.
- DEFINITION: A **file object** is a value that represents an open file and keeps track of which parts of it have been read.
- DEFINITION: A **method** is a function associated with a value (an object), called using the dot operator, as in `word.lower()`.
- DEFINITION: An **update** is an assignment that gives a new value to a variable that already exists, often computed from its old value.
- DEFINITION: To **initialize** a variable is to create it and give it a first value.
- DEFINITION: To **increment** a variable is to increase its value; to **decrement** it is to decrease its value.
- DEFINITION: An **augmented assignment operator**, such as `+=`, `-=` or `*=`, updates a variable concisely.
- DEFINITION: A **counter** is a variable used to count something, usually initialized to zero and then incremented.
- DEFINITION: A **linear search** is a pattern that goes through a sequence of elements one at a time and stops when it finds what it is looking for.
- DEFINITION: A **doctest** is a test written in a docstring: an expression after `>>>` and the value it should have.
- DEFINITION: A test **passes** if it runs and the result is as expected; otherwise it **fails**.

**Other ideas**

- Give the loop variable a name that says what it refers to, such as `letter` or `line`.
- A function that loops through a string and returns `True` or `False` can often be simplified with the `in` operator.
- Converting strings with `lower` before comparing them makes a function work with any combination of uppercase and lowercase letters.
- Doctests document what a function does and test it at the same time. When all tests pass, no news is good news.

def count_doubles(word):

    0
