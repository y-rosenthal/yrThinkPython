## How a for loop runs

The general form of a `for` loop is:

```python
for VARIABLE in ITERABLE:
    BODY
```

- `ITERABLE` is a value that gives other values, one at a time. In Chapter 3, the iterable is always a
  `range`. Later chapters loop over other iterables: the characters of a string and the lines of a file
  (Chapter 7), and the items of a list (Chapter 9).
- `VARIABLE` is the **loop variable**. Each time the loop runs, Python assigns the next value of the iterable
  to it.
- `BODY` is the indented block after the colon. One run of the body is one **iteration**.

Python runs a `for` loop in these steps:

1. Get the next value from the iterable. If there are no more values, the loop ends: go on with the code
   after the loop.
2. Assign the value to the loop variable.
3. Run the body.
4. Go back to step 1.

````python
for i in range(3):
    print('Iteration', i)
print('After the loop')
````

Step by step: `range(3)` gives the values `0`, `1` and `2`. Python assigns `0` to `i` and runs the body,
which displays `Iteration 0`. Then it assigns `1` to `i` and runs the body, then `2`. There are no more
values, so the loop ends, and Python runs the line after the loop. That line is not indented, so it is not
part of the body: it runs once.

## The loop variable

**After the loop, the loop variable keeps its last value.**

````python
for i in range(1, 4):
    print(i)
print('After the loop, i is', i)
````

**A loop over an empty range does not run its body.** `range(0)` gives no values, so the body runs zero
times, and the loop variable gets no value:

````python
for i in range(0):
    print('This line never runs')
print('Done')
````

**Each function has its own loop variable.** A loop variable is a variable like any other. When a loop is in
a function, its variable is local to the function ("Variables and parameters are local" in Chapter 3). In this
program, `print_row` and `print_box` each have a variable `i`. They are two different variables, so the loop
in `print_row` does not change the `i` of `print_box`:

````python
def print_row(width):
    line = ''
    for i in range(width):
        line = line + '*'
    print(line)

def print_box(height, width):
    for i in range(height):
        print_row(width)

print_box(3, 5)
````

`print_row` makes a string of stars: it starts with the empty string `''`, and the body adds one `'*'` in
each iteration. `print_box` calls `print_row` once in each iteration of its loop, so it displays 3 rows.
The last section of this page shows how to watch this program run in Thonny.

## Adding up values in a loop

To add up values, make a variable for the total **before** the loop, with the value `0`. In the body, add
the value of each iteration to it. After the loop, the variable has the sum:

````python
total = 0
for number in range(1, 6):
    total = total + number
print('The sum of 1 to 5 is', total)
````

`total = total + number` means: get the value of `total`, add `number`, and assign the result to `total`.
The values of `total` after each iteration are 1, 3, 6, 10 and 15.

The line `total = 0` must be before the loop. In the body, it would set the total back to 0 in each
iteration. Chapter 7 says more about updating a variable in a loop.

## Never change the iterable in the body of a loop

**The rule:** in the body of a `for` loop, do not change the iterable that the loop goes through. Do not
change a variable that the loop's `range` was made from, do not change the list that the loop goes through,
and do not assign a new value to the loop variable. Code that does this is **very bad code**. It can give a
wrong result without an error message, or never end, and a person who reads it cannot see what it does.
Never write such code. If you need a changed version of the values, make a **new** variable in the loop, as
`line` and `total` above.

The examples below show what can happen.

### Changing the variable that a range was made from

Python makes the range **once**, before the first iteration. A change to `n` in the body does not change the
range, so this loop runs 3 times, not more:

````python
n = 3
for i in range(n):
    print('i is', i, 'and n is', n)
    n = n + 1
print('After the loop, n is', n)
````

A reader who sees `range(n)` and `n = n + 1` can think that the loop never ends, or that it runs 6 times. It
runs 3 times. The code does not do what it seems to do.

### Changing the loop variable

At the start of each iteration, Python assigns the next value of the range to the loop variable. A value
that the body assigns to the loop variable is lost at the next iteration:

````python
for i in range(4):
    i = i * 10
    print(i)
````

The loop still runs 4 times, with `i` equal to 0, 1, 2 and 3 at the start of the iterations. Use a new
variable for the result instead: `ten_times = i * 10`.

### Changing a list in a loop over the list

Chapter 9 teaches **lists**. A list is a sequence of values in square brackets, for example
`[10, 20, 30, 40]`, and a `for` loop can go through its items. The method `remove` takes an item out of a
list. This loop tries to remove every item of the list, but it removes only half of them:

````python
numbers = [10, 20, 30, 40]
for n in numbers:
    print('removing', n)
    numbers.remove(n)
print(numbers)
````

The loop goes through the list by position: first position 0, then position 1, and so on. In the first
iteration, `n` is `10` (position 0), and `remove` takes it out. Now `20` is at position 0 and `30` is at
position 1. In the second iteration, the loop goes to position 1, so `n` is `30`: the loop skips `20`. After
`30` is removed, the list has two items, there is no position 2, and the loop ends. There is no error
message: the result is wrong without a warning.

Adding items is worse. The method `append` adds an item at the end of a list. In this loop, each iteration
adds an item, so the list never ends, and **the loop never ends**:

```python
numbers = [1, 2, 3]
for n in numbers:
    numbers.append(n)    # never do this: an infinite loop
```

This code is text, not a code cell, so it cannot run. If you run such a loop, stop it: in Colab, click
**Runtime**, then **Interrupt execution**; in Thonny, click the red **Stop** button.

For some iterables, Python finds the change and stops with an error. For example, a loop over a dictionary
(Chapter 10) that adds or removes items raises
`RuntimeError: dictionary changed size during iteration`.

## Watch a for loop in Thonny

one step at a time and shows the value of each variable. To install Thonny, and to learn the two parts of its

6. Press **Step out** to run the rest of the call. The window of the call closes, and Thonny goes back to the
   loop of `print_box`. Its `i` did not change: each function has its own `i`.
7. To stop the program before it ends, click the red **Stop** button (**Ctrl+F2**).

Then try the example "Changing the variable that a range was made from" in Thonny. With **Step over**, watch

### Question 6 (medium): what is wrong?

A student wants to display the numbers 0, 10, 20 and 30, and writes this code. It displays the right numbers.
Why is it bad code anyway? Write a better version.

```python
for i in range(4):
    i = i * 10
    print(i)
```

<details>
<summary>Answer</summary>

The body assigns a new value to the loop variable `i`. A reader must know that the loop gives `i` its next
value at the start of each iteration, so that the change is lost. It is easy to think that the change
affects the loop. Use a new variable instead, or a `range` that gives the right values:

```python
for i in range(4):
    print(i * 10)
```

```python
for i in range(0, 31, 10):
    print(i)
```

</details>
