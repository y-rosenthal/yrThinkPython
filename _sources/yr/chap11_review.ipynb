## Concepts covered

:::{admonition} Python Syntax and Semantics
:class: dropdown

**Creating tuples**

- A tuple is a sequence of values, indexed by integers, like a list. The values can be any type.
- The comma makes the tuple: `t = 'l', 'u', 'p', 'i', 'n'` makes a tuple of five strings.
- Parentheses are not needed, but it is common to use them: `t = ('l', 'u', 'p', 'i', 'n')` makes the same tuple.
- To make a tuple with one element, include a final comma: `t = 'p',` makes a tuple. A value in parentheses alone is not a tuple: `t = ('p')` makes the string `'p'`.
- `tuple()` makes an empty tuple: `tuple()` is `()`.
- `tuple(SEQUENCE)` makes a tuple with the elements of the sequence: `tuple('lupin')` is `('l', 'u', 'p', 'i', 'n')`.
- Do not use `tuple` as a variable name: it is the name of a built-in function.

**Indexing, slicing and operators**

- `TUPLE[INDEX]` returns the element at that index: `tuple('lupin')[0]` is `'l'`.
- `TUPLE[START:STOP]` selects a range of elements: `tuple('lupin')[1:3]` is `('u', 'p')`.
- `+` concatenates tuples: `tuple('lup') + ('i', 'n')` is `('l', 'u', 'p', 'i', 'n')`.
- `*` duplicates a tuple: `tuple('ab') * 2` is `('a', 'b', 'a', 'b')`.
- `sorted(TUPLE)` returns a list, not a tuple: `sorted(('b', 'a'))` is `['a', 'b']`.
- `reversed(TUPLE)` returns a reversed object. `tuple(reversed(TUPLE))` makes a tuple from it: `tuple(reversed(('a', 'b')))` is `('b', 'a')`.

**Tuples are immutable**

- Assigning to an element of a tuple is a `TypeError`.
- Tuples have no methods that change them, like `append` and `remove`: calling `t.remove('l')` on a tuple is an `AttributeError`.
- Because tuples are immutable, they are hashable, so they can be keys in a dictionary. After this code, `grid[1, 2]` is `3`:
  ```python
  grid = {}
  grid[1, 2] = 3
  grid[3, 4] = 7
  ```
- A tuple can be a value in a dictionary: `{'key': ('a', 'b', 'c')}` has a tuple as its value.

**Tuple assignment**

- This assigns the values to the variables from left to right:
  ```
  a, b = 1, 2
  ```
  After it, `a` is `1` and `b` is `2`.
- If the left side of an assignment is a tuple, the right side can be any kind of sequence. For example, `username, domain = email.split('@')` splits an email address at the `@`: the first part goes to `username`, the second to `domain`.
- The number of variables and the number of values must be the same, or it is a `ValueError`.
- To swap the values of two variables, use tuple assignment: `a, b = b, a`. The expressions on the right are evaluated before any of the assignments.

**Looping through a dictionary with `.items()`**

- `DICT.items()` gives the items of the dictionary, as a `dict_items` object that behaves like a list of tuples: each item is a tuple of a key and its value.
- This loops through the keys and the values together:
  ```
  for KEY, VALUE in DICT.items():
      BODY
  ```
  Each time through the loop, `KEY` and `VALUE` get the next key and its value.
- You can also loop through the items and unpack each one on its own line:
  ```python
  for item in d.items():
      key, value = item
  ```

**Tuples as return values**

- A function returns one value, but if the value is a tuple, it is like returning several values.
- `divmod(A, B)` returns a tuple of the quotient and the remainder: `divmod(7, 3)` is `(2, 1)`.
- Use tuple assignment to store the elements in variables: `quotient, remainder = divmod(7, 3)` sets `quotient` to `2` and `remainder` to `1`.
- `min(SEQUENCE)` and `max(SEQUENCE)` find the smallest and largest elements of a sequence: `min([2, 4, 1, 3])` is `1`, and `max([2, 4, 1, 3])` is `4`.

**Argument packing and unpacking**

- A parameter name that starts with `*` packs the arguments into a tuple. In `def mean(*args):`, `args` is a tuple of the arguments. `args` is the conventional name.
- `*` before a sequence unpacks it into separate arguments: `divmod(*(7, 3))` is the same as `divmod(7, 3)`.
- Passing a tuple where two arguments are expected is a `TypeError`: `divmod((7, 3))` fails, because the tuple counts as one argument.

**Zip**

- `zip(SEQUENCE, SEQUENCE)` returns a zip object, which pairs up the elements of the sequences, like the teeth of a zipper.
- This loops through the pairs:
  ```
  for pair in zip(scores1, scores2):
      BODY
  ```
  Each time through the loop, `pair` is a tuple of the two scores.
- `list(zip(SEQUENCE, SEQUENCE))` makes a list of tuples.
- `dict(zip(KEYS, VALUES))` makes a dictionary from a list of keys and a list of values: `dict(zip('ab', [1, 2]))` is `{'a': 1, 'b': 2}`.

**Enumerate**

- `enumerate(SEQUENCE)` returns an enumerate object, which loops through pairs of an index (starting from 0) and an element.
- This loops through the indices and the elements:
  ```
  for INDEX, ELEMENT in enumerate(SEQUENCE):
      BODY
  ```

**Comparing and sorting tuples**

- `<` compares tuples element by element, from the first: it stops at the first pair that differs, and later elements are not considered. `(0, 1, 2000000) < (0, 3, 4)` is `True`, because `1` is less than `3`, and the `2000000` is never compared.
- `sorted(ITEMS)` sorts a list of tuples by the first element, and uses the second element to break ties.
- `key=FUNCTION` tells `sorted` (or `max` or `min`) to use `FUNCTION` to compute the sort key for each item.
- `max(ITEMS, key=FUNCTION)` returns the item with the largest sort key. `min` works the same way for the smallest.

**Inverting a dictionary**

- To invert a dictionary, make a new dictionary where each value maps to a list of the keys that had that value: values are not unique, so one value can go with several keys, but keys are unique.
- The loop goes through the keys and the values of the old dictionary: `for key, value in d.items():`.
:::


:::{admonition} Definitions, etc.
:class: dropdown

**Definitions**

- DEFINITION: **pack** is to collect multiple arguments into a tuple.
- DEFINITION: **unpack** is to treat a tuple (or other sequence) as multiple arguments.
- DEFINITION: A **zip object** is the result of calling the built-in function `zip`, and can be used to loop through a sequence of tuples.
- DEFINITION: An **enumerate object** is the result of calling the built-in function `enumerate`, and can be used to loop through a sequence of tuples.
- DEFINITION: A **sort key** is a value, or a function that computes a value, used to sort the elements of a collection.
- DEFINITION: A **data structure** is a collection of values, organized to perform certain operations efficiently.

**Other ideas**

- A tuple that holds a mutable value, like a list, is not hashable and cannot be a key in a dictionary.
- To check the shape of a data structure, download `structshape.py` and call `structshape(VALUE)`: it returns a string that describes the structure, for example whether it is a list of tuples or a dictionary of lists.
- A tuple is a sequence, so `len`, loops and the other tools for sequences work with it.

Write a function called `make_index` that takes a list of lines and returns a dictionary. Each key is a word in the book, and its value is a list of the line numbers where the word appears. Line numbers start at 1. Each line number appears once in a list, even if the word appears more than once on that line. The method `line.split()` returns a list of the words in `line`.

Start from this header (replace `...` with the body):

```python
def make_index(lines):
    ...
```

**Examples**

```python
print(make_index(['cats sit', 'dogs sit', 'cats sleep']))
```
```text
{'cats': [1, 3], 'sit': [1, 2], 'dogs': [2], 'sleep': [3]}
```

```python
print(make_index(['one fish', 'two fish', 'red fish', 'blue fish']))
```
```text
{'one': [1], 'fish': [1, 2, 3, 4], 'two': [2], 'red': [3], 'blue': [4]}
```

:::{admonition} Answer
:class: dropdown

Loop through the lines and the words in each line. Add a word to the dictionary the first time it appears, and add the line number if it is not already there:

```python
def make_index(lines):
    index = {}
    for i in range(len(lines)):
        for word in lines[i].split():
            if word not in index:
                index[word] = []
            if i + 1 not in index[word]:
                index[word].append(i + 1)
    return index
```
Then try: `print(make_index(['cats sit', 'dogs sit', 'cats sleep']))`.

The same function with `enumerate`, which gives the position of each line (positions start at 0, so the line number is `i + 1`):

```python
def make_index(lines):

        for word in line.split():
            if word not in index:
                index[word] = []
            if i + 1 not in index[word]:
                index[word].append(i + 1)
    return index
```

For each line number in the index, build the window around it. The window starts two lines before and ends two lines after, clipped to the lines that exist:

```python
def find_in_book(lines, index, word):
    if word not in index:
        return []
    windows = []
    for number in index[word]:
        start = number - 2
        if start < 1:
            start = 1
        stop = number + 2
        if stop > len(lines):
            stop = len(lines)
        window = []
        for n in range(start, stop + 1):
            window.append((n, lines[n - 1]))
        windows.append(window)
    return windows
```

Write a function called `find_phrase` that takes a list of lines and two words, and returns a list of tuples. Each tuple is a line number and a word number for a place where the two words appear next to each other, in order, on that line. Line numbers and word numbers both start at 1. As in Question 22, the lines hold lowercase words separated by spaces, and `line.split()` returns a list of the words in `line`.

```

```

```

```

print(find_phrase(['cats sit still', 'dogs sit', 'cats sit'], 'dogs', 'cats'))
```

```

:::{admonition} Answer
:class: dropdown

Loop through the lines and the positions in each line. A phrase starts at a position where the first word matches and the next word matches the second word. The last word of a line cannot start a phrase, so the loop stops one position early:

```python
def find_phrase(lines, word1, word2):
    found = []
    for i in range(len(lines)):
        words = lines[i].split()
        for j in range(len(words) - 1):
            if words[j] == word1 and words[j + 1] == word2:
                found.append((i + 1, j + 1))
    return found
```
Then try: `print(find_phrase(['cats sit still', 'dogs sit', 'cats sit'], 'cats', 'sit'))`.
