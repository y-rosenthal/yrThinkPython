## Concepts covered

:::{admonition} Python Syntax and Semantics
:class: dropdown toggle-shown

**Defining and calling functions**

- This defines a function:
  ```
  def FUNCTION_NAME():
      BODY
  ```
  `def` is a keyword. The first line is the header and must end with a colon. The body is one or more statements, indented (by convention, four spaces).
- Defining a function does not run its body. The body runs each time the function is called: `FUNCTION_NAME()`. For example, this displays `Hello` twice:
  ```python
  def greet():
      print('Hello')

  greet()
  greet()
  ```
- A function must be defined before the line that calls it runs. Calling a name that has not been defined yet is a `NameError`.
- Any legal variable name is a legal function name.
- A function name without parentheses is a function object, not a call. In a notebook, `greet` displays `<function __main__.greet()>`, and the body does not run.

**Parameters and arguments**

- This defines a function with parameters:
  ```
  def FUNCTION_NAME(PARAMETER1, PARAMETER2):
      BODY
  ```
  `FUNCTION_NAME(ARGUMENT1, ARGUMENT2)` calls it: each argument's value is assigned to the parameter in the same position, and then the body runs.
- An argument can be any expression, such as a variable or `'a' + 'b'`. Its value is computed before the function runs.
- Calling a function with too few or too many arguments is a `TypeError`.

**Functions calling functions**

- A function's body can call other functions. When a function calls another, it waits until that call ends, then continues with its next statement.
- A function can call a function defined after it in the program, as long as both are defined before the first call runs.

**for loops**

- This runs the indented body once for each value of `range(SOME_INTEGER)`:
  ```
  for VARIABLE in range(SOME_INTEGER):
      BODY
  ```
  For example, this displays `0`, `1` and `2`, on separate lines:
  ```python
  for i in range(3):
      print(i)
  ```
- `range(SOME_INTEGER)` is the integers `0, 1, 2, ...` up to SOME_INTEGER minus 1. The loop variable takes each value in turn, even if the body does not use it.
- The header of a `for` loop ends with a colon, and its body is indented. A loop can be inside a function, and a function can be called inside a loop.

**Local variables**

- Parameters, and variables assigned inside a function, are local: they exist only while the function runs. Using one outside the function is a `NameError`.
- A local variable with the same name as a variable outside the function is a different variable: assigning to it does not change the one outside.

**Printing**

- `print()`, with no arguments, displays an empty line.
- `STRING * SOME_INTEGER` makes that many copies of the string, joined together: `'-' * 5` is `'-----'`. `' ' * 3` is three spaces.

**Tracebacks**

- When a runtime error happens inside a function, the error message includes a traceback: the function calls in progress, from the first call at the top to the function where the error happened at the bottom, ending with the error itself.
:::


:::{admonition} Definitions, etc.
:class: dropdown toggle-shown

**Definitions**

- DEFINITION: A **function definition** is a statement that creates a function.
- DEFINITION: The **header** of a function is the first line of its definition.
- DEFINITION: The **body** of a function is the sequence of statements inside its definition.
- DEFINITION: A **function object** is the value created by a function definition. The name of the function is a variable that refers to it.
- DEFINITION: A **parameter** is a name used inside a function to refer to the value passed as an argument.
- DEFINITION: A **loop** is a statement that runs one or more statements, often repeatedly.
- DEFINITION: A **local variable** is a variable defined inside a function, which can only be used inside that function.
- DEFINITION: A **stack diagram** shows the functions that are running, with their variables and the values they refer to.
- DEFINITION: A **frame** is a box in a stack diagram that represents one function call. It contains the function's parameters and local variables. Variables created outside any function are in the frame called `__main__`.
- DEFINITION: A **traceback** is the list of the functions that are running, displayed when an exception occurs.

**Other ideas**

- Reasons to write functions:
  1. A function gives a group of statements a name, which makes a program easier to read and debug.
  2. Functions make a program shorter by removing repeated code; a later change is made in one place.
  3. A long program divided into functions can be debugged one function at a time.
  4. A well-designed function can be reused in other programs.
- In a stack diagram, each function's frame is below the frame of the function that called it; a traceback lists the functions in the same order.
- Debugging is like experimental science: make a hypothesis about what is wrong, change the program, and check whether the result is what you predicted.
- Start with a working program and make small changes, testing as you go. Spending a lot of time debugging is often a sign of writing too much code before testing it.
:::

### Question 10 (medium): local variables and wrong calls

### Question 11 (medium): write a function

Write a function called `print_right` that takes a string `text` and an integer `width`, and displays `text` with enough spaces in front of it that its last character is in column `width`.
You can assume `text` has at most `width` characters.

Start from this header (replace `...` with the body):

```python
def print_right(text, width):
    ...
```

**Examples**

```python
print_right('Monty', 10)
print_right('Python', 10)
print_right('Flying Circus', 20)
```
```text
     Monty
    Python
       Flying Circus
```

```python
print_right('abc', 3)
print_right('abc', 4)
```
```text
abc
 abc
```

:::{admonition} Answer
:class: dropdown

The number of spaces is `width` minus the length of the text:

```python
def print_right(text, width):
    spaces = width - len(text)
    print(' ' * spaces + text)
```
Then try: `print_right('Monty', 10)`, `print_right('Python', 10)` and `print_right('Flying Circus', 20)`.

The same thing in one line, without the local variable `spaces`:

```python
def print_right(text, width):
    print(' ' * (width - len(text)) + text)
```

`print(' ' * spaces, text)` would not work exactly: `print` puts an extra space between its two values.

### Question 12 (medium): refactor into a function

This code displays two verses of a song. The verses are the same except for the animal and its sound.

```python
print('Old MacDonald had a farm')
print('And on that farm he had a cow')
print('With a moo moo here')
print('Old MacDonald had a farm')
print('And on that farm he had a duck')
print('With a quack quack here')
```

Write a function `verse` that takes `animal` and `sound` and displays one verse, so that the code above can be replaced by two calls: `verse('cow', 'moo')` and `verse('duck', 'quack')`.

Start from this header (replace `...` with the body):

```python
def verse(animal, sound):
    ...
```

**Examples**

```python
verse('cow', 'moo')
```
```text
Old MacDonald had a farm
And on that farm he had a cow
With a moo moo here
```

```python
verse('pig', 'oink')
```
```text
Old MacDonald had a farm
And on that farm he had a pig
With a oink oink here
```

:::{admonition} Answer
:class: dropdown

```python
def verse(animal, sound):
    print('Old MacDonald had a farm')
    print('And on that farm he had a', animal)
    print('With a', sound, sound, 'here')
```
Then try: `verse('cow', 'moo')`.

`print` separates its values with spaces, which is exactly what the lines need.
Another approach builds each line with `+` (note the spaces inside the strings):

```python
def verse(animal, sound):
    print('Old MacDonald had a farm')
    print('And on that farm he had a ' + animal)
    print('With a ' + sound + ' ' + sound + ' here')
```

Now a change to the verse (say, a new line) is made in one place instead of in every verse.

### Question 14 (hard): tracebacks

Run this cell to define `inner` and `outer`.

````python
def inner(value):
    print(value + 1)

def outer(value):
    print('outer got', value)
    inner(value)
````

Two correct ways to call it:

```python
outer(5)
```
```text
outer got 5
6
```

```python
outer(2.5)
```
```text
outer got 2.5
3.5
```

Predict what this call displays and which error happens. In the traceback, which function is listed first, and which one last?

~~~python
outer('hi')
~~~

:::{admonition} Answer
:class: dropdown

It displays `outer`'s line, and then the error happens in `inner`, when it tries to add the number `1` to the string `'hi'`: a `TypeError`.

```text
outer got hi

----> 1 outer('hi')

In outer(value):
      4 def outer(value):
      5     print('outer got', value)
----> 6     inner(value)

In inner(value):
      1 def inner(value):
----> 2     print(value + 1)
      3 
      4 def outer(value):
      5     print('outer got', value)

TypeError: can only concatenate str (not "int") to str
```

The traceback lists the calls in the order they were made: first the line `outer('hi')` in the question's code, then the line in `outer` that called `inner`, and last the line in `inner` where the error happened.
The function that was running when the error happened is at the bottom, just above the error message.

### Question 15 (hard): write a function

Write a function called `pyramid` that takes a string `char` (one character) and an integer `height`, and displays a centered pyramid with `height` lines.
Line 1 has 1 copy of `char`, line 2 has 3, line 3 has 5, and so on; each line has enough spaces in front of it to be centered above the last line.

Start from this header (replace `...` with the body):

```python
def pyramid(char, height):
    ...
```

**Examples**

```python
pyramid('*', 3)
```
```text
  *
 ***
*****
```

```python
pyramid('#', 5)
```
```text
    #
   ###
  #####
 #######
#########
```

:::{admonition} Answer
:class: dropdown

On line `i` (counting from `0`) there are `2 * i + 1` characters and `height - 1 - i` spaces:

```python
def pyramid(char, height):
    for i in range(height):
        print(' ' * (height - 1 - i) + char * (2 * i + 1))
```
Then try: `pyramid('*', 3)`.

Another approach: write a function for one line first, then call it from a loop:

```python
def pyramid_line(char, spaces, count):
    print(' ' * spaces + char * count)

def pyramid(char, height):
    for i in range(height):
        pyramid_line(char, height - 1 - i, 2 * i + 1)
```

To find the formulas, write down the number of spaces and characters for each line of an example, and look for the pattern.

### Question 16 (hard): write two functions

Write two functions that display a grid of `n` by `n` cells:

- `print_border(n)` displays one border line: `n` copies of `+---`, then a final `+`.
- `print_grid(n)` displays the whole grid. It uses `print_border`, and between two border lines it displays a line of `n` copies of `|   ` (a bar and three spaces), then a final `|`.


Start from this header (replace `...` with the body):

```python
def print_border(n):
    ...

def print_grid(n):
    ...
```

**Examples**

```python
print_border(3)
```
```text
+---+---+---+
```

```python
print_grid(2)
```
```text
+---+---+
|   |   |
+---+---+
|   |   |
+---+---+
```

```python
print_grid(3)
```
```text
+---+---+---+
|   |   |   |
+---+---+---+
|   |   |   |
+---+---+---+
|   |   |   |
+---+---+---+
```

:::{admonition} Answer
:class: dropdown

A grid of `n` rows has a border line at the top, then, for each row, a line of bars and a border line:

```python
def print_border(n):
    print('+---' * n + '+')

def print_bars(n):
    print('|   ' * n + '|')

def print_grid(n):
    print_border(n)
    for i in range(n):
        print_bars(n)
        print_border(n)
```
Then try: `print_border(3)`.

Another approach, without the helper function `print_bars`:

```python
def print_border(n):
    print('+---' * n + '+')

def print_grid(n):
    for i in range(n):
        print_border(n)
        print('|   ' * n + '|')
    print_border(n)
```
