````python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://github.com/AllenDowney/ThinkPython/raw/v3/pg345.txt');

def make_file(filename, text):
    """Make a small text file for Question 4."""
    file = open(filename, 'w')
    file.write(text)
    file.close()

make_file('sun.txt', 'The neon sun was hot.\nEve did not see the radar.\nWould you like a cold drink?\n')
make_file('noon.txt', 'At noon the level of the river was low.\n')
````

Solution 1 (a `for` loop; the loop variable is `left`):

```python
def isPalindrome(text):
    """Return True if text is a palindrome and False otherwise."""
    right = len(text) - 1
    for left in range(len(text) // 2):
        if text[left] != text[right]:
            return False
        else:
            right -= 1
    return True
```
Then try: `isPalindrome('racecar')`.

`range(len(text) // 2)` is shorthand for `0, 1, 2, ...` up to `len(text) // 2` minus 1. For `'racecar'`, `len('racecar') // 2` is `3`, so the loop runs 3 times, with `left` equal to `0`, `1` and `2`. The `for` statement moves `left` for you, so you do not write `left = 0` or `left += 1`. You must move `right` yourself (`right -= 1`).

As soon as one pair is different, `return False` ends the function. The `return True` runs only after the loop compared every pair and found no difference. It is after the loop, not in it: a `return True` inside the loop would end the function after the first pair.

Solution 2 (a `while` loop; you move both indexes yourself):

```python
def isPalindrome(text):
    """Return True if text is a palindrome and False otherwise."""
    left = 0
    right = len(text) - 1
    while left < right:
        if text[left] != text[right]:
            return False
        left += 1
        right -= 1
    return True
```

The loop stops when the two indexes meet in the middle. Both solutions return `True` for `''` and for a string of one character (the last two examples): the loop does not run, so nothing is different.

```python
def isPalindrome2(text):
    """Return True if text is a palindrome and False otherwise."""
    text_reversed = reversed(text)
    text_reversed_str = ''.join(text_reversed)
    return text_reversed_str == text
```
Then try: `isPalindrome2('racecar')`.

`''.join(...)` puts the characters together with nothing between them, so `text_reversed_str` is the reversed string. A common mistake is `str(reversed(text))`: it gives a string that starts with `'<reversed object '`, not the reversed text, so the function would always return `False`.

The last line returns the value of the comparison `text_reversed_str == text`, which is already `True` or `False`. This longer version also works:

```python
def isPalindrome2(text):
    """Return True if text is a palindrome and False otherwise."""
    text_reversed_str = ''.join(reversed(text))
    if text_reversed_str == text:
        return True
    else:
        return False
```

But it is too wordy: `return text_reversed_str == text` is shorter and gives the same results. Use that form when a function returns the result of a comparison.

```python
def isPalindrome(text):
    """Return True if text is a palindrome and False otherwise."""
    right = len(text) - 1
    for left in range(len(text) // 2):
        if text[left] != text[right]:
            return False
        else:
            right -= 1

        if isPalindrome(word):
            answer.append(word)
```

```python
def findPalindromesInFile(filename):
```

```python
    answer = []
    for line in open(filename, encoding='utf-8'):

    return answer
```

### Question 5 (hard): write a recursive function

Write a **recursive** version of the test, `isPalindrome3`: a function that calls itself.

The idea: a string is a palindrome if its first and last characters are the same **and** the string between them is a palindrome. For example, `'racecar'` is a palindrome because `'r'` equals `'r'` and `'aceca'` is a palindrome; `'aceca'` is a palindrome because `'a'` equals `'a'` and `'cec'` is a palindrome; and so on. A string with 0 or 1 characters is always a palindrome: that is the base case, where the function stops calling itself.

Hint: `text[0]` is the first character, `text[-1]` is the last character, and `text[1:-1]` is the string without its first and last characters: `'racecar'[1:-1]` is `'aceca'`.

Start from this header (replace `...` with the body):

```python

    ...
```

**Examples**

```text
True
```

```python

```
```text
False
```

```python

```
```text

```

:::{admonition} Answer
:class: dropdown

```python
def isPalindrome3(text):
    """Return True if text is a palindrome and False otherwise."""
    if len(text) <= 1:
        return True
    if text[0] != text[-1]:
        return False
    return isPalindrome3(text[1:-1])
```
Then try: `isPalindrome3('racecar')`.

The function has the two parts of every recursive function:

- **The base case:** `len(text) <= 1`. The function returns without calling itself. Without a base case, the calls would never stop (Python stops them with a `RecursionError`). The base case must cover the empty string too: `'noon'` gets shorter by two characters with each call, `'noon'[1:-1]` is `'oo'` and `'oo'[1:-1]` is `''`, so it never has exactly one character.
- **The recursive case:** the function calls itself with a shorter string, `text[1:-1]`. Each call is closer to the base case.

Note the word `return` in the last line. Without it, the function would call itself, ignore the result, and return `None`.

For `'racebar'` the calls are `isPalindrome3('racebar')`, then `isPalindrome3('aceba')` (`'racebar'[1:-1]` is `'aceba'`); there `'a'` and `'a'` are the same, so it calls `isPalindrome3('ceb')` (`'aceba'[1:-1]` is `'ceb'`), where `'c'` and `'b'` are different, so it returns `False`. Each call returns that `False` to the call before it.
