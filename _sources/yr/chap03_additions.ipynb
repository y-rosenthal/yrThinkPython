## The three forms of `range`

Chapter 3 uses `range` with one argument: `range(SOME_INTEGER)` gives the integers `0, 1, 2, ...` up to
SOME_INTEGER minus 1. For example, this loop displays `0`, `1` and `2`:

````python
for i in range(3):
    print(i)
````

`range` can also take two or three arguments:

| Form | The values it gives |
|---|---|
| `range(STOP)` | `0, 1, 2, ...` up to STOP minus 1 |
| `range(START, STOP)` | `START, START + 1, START + 2, ...` up to STOP minus 1 |
| `range(START, STOP, STEP)` | `START, START + STEP, START + 2*STEP, ...` while the value is before STOP |

In all three forms, **STOP itself is never one of the values.** The sections below show each form.

## `range(START, STOP)`

With two arguments, the first argument is the value to start at, and the second is the value to stop
before. This loop displays the integers from `3` to `6` (it stops before `7`):

````python
for i in range(3, 7):
    print(i)
````

`range(STOP)` is the same as `range(0, STOP)`: when you do not give a start, the start is `0`.

`range(START, STOP)` gives STOP minus START values. So `range(3, 7)` gives 4 values.

If START is equal to STOP or greater than STOP, the range is empty, and the body of the loop does not run.
Here, the loop displays nothing; only the last line displays something:

````python
for i in range(5, 5):
    print(i)
print('The loop is done.')
````

To include STOP in the values, use STOP plus 1 as the second argument. This loop displays `1`
to `5`:

````python
for i in range(1, 5 + 1):
    print(i)
````

The arguments can be any expressions that have an integer value, such as variables and parameters.
This function displays the integers from `low` to `high`, including `high`:

````python
def print_numbers(low, high):
    for i in range(low, high + 1):
        print(i)

print_numbers(8, 11)
````

## `range(START, STOP, STEP)`

With three arguments, the third argument is the *step*: the amount to add to get the next value.
This loop displays the even numbers from `0` to `8`:

````python
for i in range(0, 10, 2):
    print(i)
````

The last value is `8`, because the next value, `10`, is not before STOP.
The last value is not always STOP minus 1. Here, it is `7`:

````python
for i in range(1, 10, 3):
    print(i)
````

`range(START, STOP)` is the same as `range(START, STOP, 1)`: when you do not give a step,
the step is `1`.

## Counting down: a negative step

When STEP is negative, the values go down, and the range stops before it gets to STOP.
This loop counts down from `5` to `1`:

````python
for i in range(5, 0, -1):
    print(i)
print('Liftoff!')
````

Chapter 3 uses this form to count down from `99` to `1`: `range(99, 0, -1)`.

To include `0`, the stop must be one step past it, which is `-1`. This loop displays `3`, `2`, `1` and `0`:

````python
for i in range(3, -1, -1):
    print(i)
````

With a negative step, START must be greater than STOP. If it is not, the range is empty.
This loop displays nothing, because you cannot count *down* from `1` to `5`:

````python
for i in range(1, 5, -1):
    print(i)
print('The loop is done.')
````

## Arguments that `range` does not accept

The arguments of `range` must be integers. A float, even one like `2.0`, causes a `TypeError`:

````python

for i in range(2.0):
    print(i)
````

The step cannot be `0`, because the values would never get to STOP. A step of `0` causes a
`ValueError`:

````python

for i in range(1, 10, 0):
    print(i)
````

### Question 6 (medium): write a function

Write a function called `count_down` that takes an integer `n` and displays the integers from `n` down
to `1`, each on its own line, and then displays `Go!`. For example, `count_down(3)` displays:

```text
3
2
1
Go!
```

Start from this header (replace `...` with the body):

```python
def count_down(n):
    ...
```

:::{admonition} Answer
:class: dropdown

```python
def count_down(n):
    for i in range(n, 0, -1):
        print(i)
    print('Go!')
```

The stop is `0`, so the last value is `1`. The call `print('Go!')` is not indented under the `for`
header, so it runs once, after the loop.

### Question 7 (hard): write a function

Write a function called `print_multiples` that takes two integers, `n` and `count`, and displays the first
`count` multiples of `n` (`n`, `2*n`, `3*n`, ...), each on its own line. For example,
`print_multiples(7, 4)` displays `7`, `14`, `21` and `28`.

Use a `range` with three arguments. Start from this header (replace `...` with the body):

```python
def print_multiples(n, count):
    ...
```

:::{admonition} Answer
:class: dropdown

```python
def print_multiples(n, count):
    for i in range(n, n * count + 1, n):
        print(i)
```

The range starts at `n` and adds `n` each time. The last multiple is `n * count`; the stop is one more
than that, so that the last multiple is included.

Another way uses a range with one argument and multiplies:

```python
def print_multiples(n, count):
    for i in range(count):
        print(n * (i + 1))
```

### Question 8 (hard): find the error

This function should display the integers from `high` down to `low`, including both. For example,
`print_down(5, 2)` should display `5`, `4`, `3` and `2`. But it displays nothing. Explain why, and fix it.

````python
def print_down(high, low):
    for i in range(high, low):
        print(i)

print_down(5, 2)
````

:::{admonition} Answer
:class: dropdown

The range has no step, so the step is `1`, and the range counts up from `high`. Because `high` is greater
than `low`, the range is empty and the loop body never runs.

The fix uses a step of `-1`, and a stop one step past `low`, so that `low` is included:

```python
def print_down(high, low):
    for i in range(high, low - 1, -1):
        print(i)
```
:::
