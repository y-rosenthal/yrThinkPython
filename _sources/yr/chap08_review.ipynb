## Concepts covered

:::{admonition} Python Syntax and Semantics
:class: dropdown toggle-shown

**Indexing**

- A string is a sequence of characters. `STRING[INDEX]` selects the character at position INDEX. The first character has index `0`: if `fruit = 'banana'`, then `fruit[0]` is `'b'` and `fruit[1]` is `'a'`.
- The index can be any expression whose value is an integer, such as `fruit[i+1]`. An index that is not an integer, such as `fruit[1.5]`, is a `TypeError`.
- `len(STRING)` returns the number of characters in the string. The indices of a string go from `0` to `len(STRING) - 1`. An index after the last character, such as `fruit[len(fruit)]`, is an `IndexError`.
- A negative index counts backward from the end: `STRING[-1]` is the last character, `STRING[-2]` is the second to last, and so on. `fruit[-1]` is `'a'` and `fruit[-2]` is `'n'`.
- This runs the body once for each index of the string, with `i` referring to `0`, `1`, `2`, ... up to the length minus 1:
  ```
  for i in range(len(STRING)):
      BODY
  ```
  For example, this displays `0 b`, `1 a` and so on, on separate lines:
  ```python
  for i in range(len(fruit)):
      print(i, fruit[i])
  ```

**Slices**

- `STRING[N:M]` returns the part of the string from index N up to, but not including, index M: `fruit[0:3]` is `'ban'` and `fruit[3:6]` is `'ana'`.
- If you omit the first index, the slice starts at the beginning of the string: `fruit[:3]` is `'ban'`. If you omit the second index, the slice goes to the end of the string: `fruit[3:]` is `'ana'`. `fruit[:]` is the whole string.
- If the first index is greater than or equal to the second, the slice is the **empty string**, `''`, which has length `0`: `fruit[3:3]` is `''`.
- An index after the end of the string is not an error in a slice: the slice stops at the end of the string. `fruit[3:100]` is `'ana'`.
- Negative indices also work in slices: `fruit[:-1]` is `'banan'` (all the characters except the last).

**Strings are immutable**

- A string cannot be changed. An assignment to a character of a string, such as `fruit[0] = 'B'`, is a `TypeError`.
- To get a variation of a string, create a new string, for example with `+` and a slice: `'B' + fruit[1:]` is `'Banana'`.
- String methods never change the string they are invoked on: they return a new string.

**Comparing strings**

- `==` and `!=` check whether two strings are equal. Uppercase and lowercase letters are different: `'Cat' == 'cat'` is `False`.
- `<`, `>`, `<=` and `>=` compare strings in alphabetical order: `'apple' < 'banana'` is `True`.
- All the uppercase letters come before all the lowercase letters: `'Pineapple' < 'banana'` is `True`. To compare words in alphabetical order, convert them to lowercase first.
- A string that is the beginning of a longer string comes before it: `'ban' < 'banana'` is `True`.

**String methods**

- A method is invoked with the dot operator: `STRING.METHOD_NAME(ARGUMENTS)`.
- `STRING.upper()` returns a new string with all the letters in uppercase: `'banana'.upper()` is `'BANANA'`. (`STRING.lower()` does the same with lowercase.)
- `STRING.startswith(PREFIX)` returns `True` if STRING starts with PREFIX, otherwise `False`: `'banana'.startswith('ban')` is `True`. `STRING.endswith(SUFFIX)` checks the end of the string in the same way: `'banana'.endswith('ban')` is `False`.
- `STRING.count(PART)` returns the number of times PART appears in STRING: `'banana'.count('a')` is `3`. The parts that it counts do not overlap: `'banana'.count('ana')` is `1`.
- `STRING.replace(OLD, NEW)` returns a new string in which every OLD is replaced with NEW: `'banana'.replace('a', 'o')` is `'bonono'`.
- `PART in STRING` is `True` if PART appears anywhere in STRING: `'nan' in 'banana'` is `True`.

**Writing files**

- `open(FILE_NAME)` opens a file for reading. `open(FILE_NAME, 'w')` opens a file for writing: if the file does not exist, it is created; if it exists, its contents are replaced.
- `FILE_OBJECT.write(STRING)` writes the string to the file. It does not add a newline: to end a line, put `'\n'` at the end of the string.
- `FILE_OBJECT.close()` closes the file. Close a file when you are done writing it: what you wrote is not certain to be in the file until it is closed.
- A file object keeps track of where it is in the file. A second loop over the same file object continues after the last line that the first loop read. To start again at the beginning, open the file again.

**The break statement**

- `break` inside a loop ends the loop immediately. The program continues with the first statement after the loop, and the loop variable keeps the value it had when the loop ended:
  ```
  for LOOP_VARIABLE in SEQUENCE:
      if CONDITION:
          break
      BODY
  ```
  For example, this displays `b` and `a`, and stops at the first `n`:
  ```python
  for letter in 'banana':
      if letter == 'n':
          break
      print(letter)
  ```

**Regular expressions**

- `import re` makes the functions of the module `re` (regular expressions) available.
- `re.search(PATTERN, STRING)` returns a `Match` object if the pattern matches some part of the string, and `None` if it does not. To check, compare the result with `None`: `re.search('nan', 'banana') != None` is `True`.
- If `result` is a `Match` object, `result.string` is the string that was searched, `result.group()` returns the part of the string that matched, and `result.span()` returns two numbers in parentheses: the index where the match starts and the index just after it ends. `re.search('nan', 'banana').span()` is `(2, 5)`.
- In a pattern, letters, digits and spaces match themselves. These characters are special:
  - `|` matches either the pattern on its left or the pattern on its right: `'Mina|Murray'` matches `Mina` or `Murray`.
  - `^` matches the beginning of the string: `'^Dracula'` matches a string that starts with `Dracula`.
  - `$` matches the end of the string (before a newline at the end): `'Harker$'` matches a string that ends with `Harker`.
  - Parentheses enclose the part of the pattern that `|` applies to: `'cent(er|re)'` matches `center` and `centre`.
  - `?` makes the character before it optional: `'colou?r'` matches `color` and `colour`.
- `re.sub(PATTERN, REPLACEMENT, STRING)` returns a new string in which every part of STRING that matches PATTERN is replaced with REPLACEMENT: `re.sub('colou?r', 'color', 'colour and color')` is `'color and color'`.
- A regular expression is often written as a raw string, with the letter `r` before the quotation mark, such as `r'colou?r'`.

**Shell commands**

- In Jupyter and Colab, a line that starts with `!` is a shell command, not Python. `!head FILE_NAME` displays the first 10 lines of a file, `!head -100 FILE_NAME` displays the first 100, and `!tail FILE_NAME` displays the last 10. `>` writes the result to a file instead of displaying it: `!head pg345.txt > first_lines.txt`. These commands are not available on all operating systems.
:::


:::{admonition} Definitions, etc.
:class: dropdown toggle-shown

**Definitions**

- DEFINITION: A **sequence** is an ordered collection of values, in which each value is identified by an integer index.
- DEFINITION: A **character** is an element of a string: a letter, a digit, a punctuation mark or white space.
- DEFINITION: An **index** is an integer that selects an item in a sequence, such as a character in a string. In Python, indices start at `0`.
- DEFINITION: A **slice** is a part of a string specified by a range of indices.
- DEFINITION: The **empty string**, `''`, is a string that contains no characters and has length `0`.
- DEFINITION: An **object** is something a variable can refer to. An object has a type and a value.
- DEFINITION: An object is **immutable** if its elements cannot be changed. Strings are immutable.
- DEFINITION: An **invocation** is an expression, or part of an expression, that calls a method, as in `word.upper()`.
- DEFINITION: A **regular expression** is a sequence of characters that defines a search pattern.
- DEFINITION: A **pattern** is a rule that specifies the requirements a string has to meet to be a match.
- DEFINITION: **String substitution** is the replacement of a string, or part of a string, with another string.
- DEFINITION: A **shell command** is a statement in a shell language, a language used to interact with the operating system.

**Other ideas**

- Indices start at `0`, so the last index of a string is its length minus 1.
- Convert strings to a standard format, such as all lowercase, before comparing them.
- If you know exactly which characters to look for, the `in` operator and the `replace` method are enough. Regular expressions can do more: match one of several words, the beginning or the end of a string, or an optional character.
- When a program works with a large file, start with a small part of the file (for example its first 10 or 100 lines), get the program working, and then run it with the whole file.
:::

:::{admonition} Answer
:class: dropdown

Two slices: the part from index `n` to the end, then the first `n` letters:

```python
def rotate(word, n):

```
Then try: `print(rotate('python', 2))`.

Another way: a loop that adds the letters one at a time. The letter at index `i` of the result is the letter at index `i + n` of `word`; the `%` operator makes the index go back to `0` after the end of the word:

```python
def rotate(word, n):
    result = ''
    for i in range(len(word)):
        result += word[(i + n) % len(word)]
    return result
```

`re` is the module for regular expressions; it was imported with `import re`.

~~~python
text = 'My favourite colour is the colour of the sea.'
print(re.sub('colou?r', 'color', text))
print(re.sub('favou?rite', 'favorite', text))
print(re.sub('^My', 'Your', text))
print(re.sub('sea$', 'sky', text))
print(text)
~~~

:::{admonition} Answer
:class: dropdown

```text
My favourite color is the color of the sea.
My favorite colour is the colour of the sea.
Your favourite colour is the colour of the sea.
My favourite colour is the colour of the sea.
My favourite colour is the colour of the sea.
```

`re.sub` replaces every part of the string that matches the pattern, so both `colour`s become `color`. Each call starts from the original `text`, so the first line does not change `favourite`.
`'^My'` matches `My` at the beginning of the string.
`'sea$'` does not match anything: `text` ends with `sea.`, and the period is the last character. So `re.sub` returns the string unchanged.
Strings are immutable, so `re.sub` returns new strings and `text` does not change.
:::


### Question 17 (hard): what is displayed?

~~~python
writer = open('story.txt', 'w')
writer.write('Title\n*** START\nOnce upon a time\nThe end\n*** END\nLicense\n')
writer.close()

reader = open('story.txt')
for line in reader:
    if line.startswith('*** '):
        break
for line in reader:
    if line.startswith('*** '):
        break
    print(line.strip())
print(line.strip())
reader.close()
~~~

:::{admonition} Answer
:class: dropdown

```text
Once upon a time
The end
*** END
```

The file has 6 lines. The first loop reads `Title` and `*** START`, and `break` ends it at `*** START`.
`reader` keeps track of where it is in the file, so the second loop continues with the next line, `Once upon a time`. It displays each line until it gets to `*** END`, and `break` ends it there.
After the loop, `line` still refers to `'*** END\n'`, so the last `print` displays it. The line `License` is never read.

### Question 18 (hard): write a function

A *palindrome* is a word that reads the same backward and forward, such as `racecar`.
Write a function called `is_palindrome` that takes a string and returns `True` if it is a palindrome and `False` otherwise. Uppercase and lowercase letters count as the same letter.

Start from this header (replace `...` with the body):

```python
def is_palindrome(word):
    ...
```

**Examples**

```python
print(is_palindrome('racecar'), is_palindrome('Noon'))
```
```text
True True
```

```python
print(is_palindrome('python'), is_palindrome('a'), is_palindrome(''))
```
```text
False True True
```

:::{admonition} Answer
:class: dropdown

Compare each letter with the letter at the same distance from the end. As soon as two letters are different, the answer is `False`:

```python
def is_palindrome(word):
    word = word.lower()
    for i in range(len(word)):
        if word[i] != word[-1 - i]:
            return False
    return True
```
Then try: `print(is_palindrome('racecar'), is_palindrome('Noon'))`.

Or make the reversed string and compare it with the word:

```python
def is_palindrome(word):
    word = word.lower()
    backward = ''
    for letter in word:
        backward = letter + backward
    return word == backward
```

Both versions call `lower` first, so that `N` and `n` are the same letter.
The first version checks each pair of letters twice; a loop over `range(len(word) // 2)` would be enough.
:::


### Question 19 (hard): write a function

Write a function called `count_words` that takes a regular expression `pattern` and returns the number of words in the file `words.txt` that the pattern matches.
`words.txt` has one word per line, in lowercase.

Start from this header (replace `...` with the body):

```python
def count_words(pattern):
    ...
```

**Examples**

```python
print(count_words('^q'), count_words('ology$'))
```
```text
568 100
```

```python
print(count_words('^(un|re)'), count_words('colou?r'))
```
```text
6614 57
```

This question uses `has_match` from Question 15.

:::{admonition} Answer
:class: dropdown

A counter, and a loop through the lines of the file:

```python
import re

def count_words(pattern):
    count = 0
    for line in open('words.txt'):
        word = line.strip()
        if re.search(pattern, word) != None:
            count += 1
    return count
```
Then try: `print(count_words('^q'), count_words('ology$'))`.

Each line is stripped first, so that the newline is not part of the word.
With the function `has_match` from Question 15, the test is shorter:

```python
def count_words(pattern):
    count = 0
    for line in open('words.txt'):
        if has_match(pattern, line.strip()):
            count += 1
    return count
```

### Question 20 (hard): write a function

Write a function called `wordle_feedback` that takes two strings of the same length, `target` and `guess`, and returns a string with one character for each letter of `guess`:
`G` if the letter is at the same position in `target`, `Y` if it is somewhere else in `target`, and `-` if it is not in `target`.

Start from this header (replace `...` with the body):

```python
def wordle_feedback(target, guess):
    ...
```

**Examples**

```python
print(wordle_feedback('mower', 'tried'))
```
```text
-Y-G-
```

```python
print(wordle_feedback('spade', 'spade'), wordle_feedback('spade', 'clerk'))
```
```text
GGGGG --Y--
```

**By the way:** this is the information that the word game Wordle gives after each guess. (The real game handles letters that appear twice in a different way.)

:::{admonition} Answer
:class: dropdown

Loop through the indices, so that the letters of `guess` and `target` at the same position can be compared:

```python
def wordle_feedback(target, guess):
    result = ''
    for i in range(len(guess)):
        if guess[i] == target[i]:
            result += 'G'
        elif guess[i] in target:
            result += 'Y'
        else:
            result += '-'
    return result
```
Then try: `print(wordle_feedback('mower', 'tried'))`.

Or loop through the letters of `guess`, and keep the index in a variable of your own:

```python
def wordle_feedback(target, guess):
    result = ''
    i = 0
    for letter in guess:
        if letter == target[i]:
            result += 'G'
        elif letter in target:
            result += 'Y'
        else:
            result += '-'
        i += 1
    return result
```

The order of the tests is important: check for `G` first, because a letter at the right position is also in `target`.
