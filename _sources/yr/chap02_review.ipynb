## Concepts covered

:::{admonition} Python Syntax and Semantics
:class: dropdown toggle-shown

**Variables and assignment**

- `VARIABLE_NAME = EXPRESSION` computes the value of the expression and makes the variable refer to it. For example, `n = 17` creates the variable `n` with the value `17`.
- The variable goes on the left of `=`: `17 = n` is a `SyntaxError`.
- An assignment statement displays nothing.
- A variable can be used anywhere its value can: in expressions and as a function argument. For example, after `n = 17`, `n + 25` is `42` and `print(n)` displays `17`.
- Assigning to an existing variable makes it refer to the new value. A variable computed from it earlier keeps its old value: after `x = 5`, `y = x + 1` and `x = 10`, `y` is still `6`.
- `NAME1 = NAME2 = EXPRESSION` assigns the same value to both variables.
- Using a variable that has not been assigned is a `NameError`.

**Variable names**

- A name can contain letters, digits and the underscore, `_`, and cannot start with a digit. No spaces or other punctuation are allowed. For example, `your_name` and `score2` are legal; `2nd_place`, `million!` and `my score` are a `SyntaxError`.
- By convention, variable names use only lowercase letters, with `_` between words.
- Uppercase and lowercase letters are different: `Score` and `score` are two different names.
- A keyword (a word such as `class`, `def`, `for`, `import`, `if`, `lambda`, `True`, `None`) cannot be a variable name: `class = 'Python'` is a `SyntaxError`.
- `int`, `float` and `str` are not keywords, so they are legal variable names, but using them as names is a bad idea.

**Importing modules**

- `import MODULE_NAME` makes a module's variables and functions available as `MODULE_NAME.NAME`. For example, after `import math`, `math.pi` is `3.141592653589793`.
- The math module provides `math.pi`, `math.e`, `math.sqrt(X)` (square root), `math.pow(X, Y)` (X to the power Y), `math.exp(X)` (e to the power X), `math.sin(X)` and `math.cos(X)`. These functions return floats: `math.sqrt(25)` is `5.0` and `math.pow(5, 2)` is `25.0`.
- `X ** Y` also raises X to the power Y, and is used more often than `math.pow`. With two integers it produces an integer: `5 ** 2` is `25`.
- Importing a module that does not exist, such as `import maath`, is a `ModuleNotFoundError`.
- Using `math.pi` without first running `import math` is a `NameError`.

**Expressions, statements and display**

- An expression (a value, a variable, a function call, or a combination of these with operators) has a value. A statement (such as an assignment or an import) has an effect but no value.
- In a notebook cell, only the value of the last expression is displayed.
- `print(VALUE1, VALUE2, ...)` displays any number of values on one line, separated by spaces. For example, `print('The answer is', 42)` displays `The answer is 42`.
- Each call to `print` displays a new line. `print` displays strings without their quotes.

**Function arguments**

- Some functions take one argument, like `len(STRING)`; some take two, like `math.pow(X, Y)`; some take any number, like `print`.
- Some arguments are optional. `int(STRING, BASE)` reads the string as a number in that base: `int('101', 2)` is `5`. `round(NUMBER, DIGITS)` rounds to that many decimal places: `round(3.14159, 3)` is `3.142`.
- Calling a function with too many arguments, too few arguments, or an argument of the wrong type is a `TypeError`. For example, `math.pow(2)` and `math.sqrt('123')` are each a `TypeError`.

**Comments**

- `#` starts a comment: Python ignores everything from the `#` to the end of the line. A comment can be on its own line or at the end of a line of code.
- A `#` inside a string is part of the string, not a comment: `print('a # b')` displays `a # b`.

**Errors**

- If a cell has a syntax error anywhere, none of its code runs: Python displays the error immediately.
- A runtime error stops the program at the line where it happens; the lines before it have already run.
:::


:::{admonition} Definitions, etc.
:class: dropdown toggle-shown

**Definitions**

- DEFINITION: A **variable** is a name that refers to a value.
- DEFINITION: An **assignment statement** is a statement that assigns a value to a variable.
- DEFINITION: A **state diagram** shows a set of variables, each name with an arrow pointing to its value.
- DEFINITION: A **keyword** is a special word used to specify the structure of a program.
- DEFINITION: A **module** is a file that contains Python code, such as variables and function definitions.
- DEFINITION: An **import statement** is a statement that reads a module so its variables and functions can be used.
- DEFINITION: The **dot operator**, `.`, is used to get a variable or function from a module, as in `math.sqrt`.
- DEFINITION: To **evaluate** an expression is to compute its value.
- DEFINITION: A **statement** is a unit of code that has an effect but no value, such as an assignment or an import.
- DEFINITION: To **execute** a statement is to run it and do what it says.
- DEFINITION: An **argument** is a value given to a function when it is called.
- DEFINITION: A **comment** is text in a program that explains the program and has no effect on how it runs.
- DEFINITION: A **syntax error** is an error in the structure of a program.
- DEFINITION: A **runtime error** is an error that makes a running program display an error message and stop.
- DEFINITION: An **exception** is an error detected while the program is running (another name for a runtime error).
- DEFINITION: A **semantic error** is an error that makes a program do the wrong thing without displaying an error message.

**Other ideas**

- Comments are most useful when they explain what is not obvious from the code, especially *why* the code does something. A comment that repeats what the code says is useless.
- Good variable names reduce the need for comments, but very long names make expressions hard to read.
- Knowing which kind of error you have (syntax, runtime or semantic) helps you find it faster.
- When you learn a new feature, make mistakes on purpose to learn what the error messages mean.
:::

### Question 10 (medium): wrong calls

`math.pow(X, Y)` raises X to the power Y. Two correct ways to call it:

```python
import math
print(math.pow(5, 2))
```
```text
25.0
```

```python
import math
print(math.pow(2, 0.5))
```
```text

```

Each of the next three calls is wrong in a different way. Predict the error for each, then open the Answer to check.

**Part a**

~~~python
import math
math.pow(2)
~~~

**Part b**

~~~python
import math
math.pow(2, 3, 4)
~~~

**Part c**

~~~python
import math
math.pow('2', 3)
~~~

:::{admonition} Answer
:class: dropdown

**Part a:** Too few arguments (a `TypeError`):

```text
      1 import math
----> 2 math.pow(2)

TypeError: pow expected 2 arguments, got 1
```

**Part b:** Too many arguments (a `TypeError`):

```text

----> 2 math.pow(2, 3, 4)

```

**Part c:** `'2'` is a string, not a number. Its quotes make it a string even though it contains a digit (a `TypeError`):

```text

----> 2 math.pow('2', 3)

```

All three are a `TypeError`. Use `math.pow(2, 3)` (or `2 ** 3`) instead.

### Question 15 (hard): find the error

This code should display the average of three test scores, which is `80.0`.
What does it display instead? What kind of error is it (syntax, runtime or semantic), and how do you fix it?

~~~python
score1 = 90
score2 = 80
score3 = 70
average = score1 + score2 + score3 / 3
print(average)
~~~

:::{admonition} Answer
:class: dropdown

```text
193.33333333333334
```

Division happens before addition, so only `score3` is divided by 3: `90 + 80 + 23.333333333333332`.
It is a **semantic error**: the code runs without an error message, but does not do what was intended.
Fix it with parentheses:

```python
average = (score1 + score2 + score3) / 3
print(average)
```
```text
80.0
```
:::

Write code that displays the total cost, `price` times `quantity`, rounded to 2 decimal places, in the form shown below.

### Question 17 (hard): write code

The variable `total` holds a number of seconds (a whole number).
Write code that displays it as hours, minutes and seconds, in the form shown below.
Use integer division, `//`, which divides and rounds down to an integer (for example, `7 // 2` is `3`).

**Examples**

Run this cell to assign `total`.

````python
total = 3725
````

```text
1 h 2 min 5 s
```
```python
total = 7384
```
```text
2 h 3 min 4 s
```

:::{admonition} Answer
:class: dropdown

Take out the whole hours first, then the whole minutes from what is left:

```python
hours = total // 3600              # 3600 seconds in an hour
rest = total - hours * 3600        # seconds left after the whole hours
minutes = rest // 60
seconds = rest - minutes * 60
print(hours, 'h', minutes, 'min', seconds, 's')
```

Another approach computes the minutes from the total first:

```python
total_minutes = total // 60
seconds = total - total_minutes * 60
hours = total_minutes // 60
minutes = total_minutes - hours * 60
print(hours, 'h', minutes, 'min', seconds, 's')
```

**By the way:** Python also has an operator, `%`, that computes the remainder of a division directly (`3725 % 3600` is `125`), which makes this shorter.
