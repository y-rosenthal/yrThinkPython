## The game

Othello is a game for two players, **Black** and **White**, on a board of 8 rows and 8 columns. The columns
have the letters `a` to `h`, and the rows have the numbers `1` to `8`, so each square has a name such as `d3`
(column `d`, row `3`). Each player puts discs of their color on the board: `B` for Black and `W` for White.

**The start.** The game starts with 4 discs in the middle: White on `d4` and `e5`, Black on `e4` and `d5`.
Black moves first. Then the players take turns.

**A move.** A player puts one disc of their color on an empty square. The move is **legal** only if it
**outflanks** at least one disc of the other color: in at least one of the 8 directions from the new disc (up,
down, left, right, and the 4 diagonals), there is a line of one or more discs of the other color, and right
after them a disc of the player's color. All the discs of the other color in each such line **flip**: they get
the player's color. A move can flip discs in several directions at once.

For example, at the start, Black can play `d3`. Down from `d3` there is the white disc on `d4`, and right after
it the black disc on `d5`. So the move outflanks `d4`, and `d4` flips to black (see the first example run below).
Black cannot play `a1`: no line of white discs goes from `a1` to a black disc.

**Passing.** If a player has no legal move, they must pass, and the other player moves again. If a player has
a legal move, they must move.

**The end.** The game ends when neither player has a legal move (usually when the board is full). The player
with more discs on the board wins. With the same number of discs, the game is a tie.

Write a program in which two people play Othello at the same keyboard. The program must do these things:

1. Show the board as in the example runs: a line with the letters of the columns, then each row with its
   number. An empty square is `.`, a black disc `B`, a white disc `W`. An empty line follows the board.
2. Before each move, show the board, and the number of discs of each color: `Black: 2  White: 2`.
3. Ask the player whose turn it is for a move: `Black (B), your move: `. A move is the name of a square, such
   as `d3`.
4. If the reply is not a letter `a` to `h` followed by a number `1` to `8`, ask again with
   `Type a column letter and a row number, for example d3: `. If the square is not a legal move, ask again with
   `a1 is not a legal move. Try again: ` (with the square that the player typed).
5. Put the disc on the board, and flip the outflanked discs in every direction.
6. If the player whose turn it is has no legal move, display `Black has no legal move and must pass.` (or
   `White ...`), and go on with the other player.
7. When neither player has a legal move, show the board, then `Game over. Black: 26  White: 38` (with the
   numbers of discs), then `Black wins.`, `White wins.` or `It is a tie.`
