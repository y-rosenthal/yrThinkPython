Write a function called `word_lengths` that takes a list of strings and returns a new list with the length of each string.

Start from this header (replace `...` with the body):

```python

    ...
```

**Examples**

```python

```


```text

```

:::{admonition} Answer
:class: dropdown

```python
def word_lengths(words):
    result = []
    for w in words:
        result.append(len(w))
    return result
```
Then try: `word_lengths(['hello', 'world'])`.

This is the accumulator pattern with a list: start with an empty list, and `append` one value in each pass of the loop. For an empty list, the loop does not run, and the function returns `[]`.

Solution 1 (build a new string, one character at a time):

```python
def remove_chars(text, chars):
    result = ''
    for ch in text:
        if ch not in chars:
```

```python
            result = result + new
            skip = m - 1
        else:
            result = result + s[i]
    return result
```
Then try: `my_replace('hello', 'l', 'L')`.

A `for` loop moves one index at a time. After a match at index `i`, the characters at indexes `i + 1` to `i + m - 1` are part of the match, so the next `m - 1` passes of the loop do nothing (`skip`). At the end of the string, the slice `s[i:i + m]` is shorter than `m` characters, so it does not match.

Solution 2 (split and join):

```python
def my_replace(s, old, new):
    return new.join(s.split(old))
```

`s.split(old)` returns the parts of `s` between the occurrences of `old`, and `new.join(...)` puts the parts together again with `new` between them.

### Question 36 (hard): write a function

Write a function called `my_strip` that takes a string and returns it without the spaces at the start and at the end. Do what `s.strip()` does for spaces, but do not use `strip`.

Hint: with a loop, find the index of the first character that is not a space, and the index of the last one. Then return one slice.

Start from this header (replace `...` with the body):

```python
    ...
```

**Examples**

```python

```
```text
'hello'
```

```python
my_strip('  two words ')
```
```text
'two words'
```

```python
my_strip('   ')
```
```text
''
```

```python
my_strip('x')
```
```text
'x'
```

:::{admonition} Answer
:class: dropdown

```python
def my_strip(s):
    start = -1
    end = -1
    for i in range(len(s)):
        if s[i] != ' ':
            if start == -1:
                start = i
            end = i
    if start == -1:
        return ''
    return s[start:end + 1]
```
Then try: `my_strip('   hello   ')`.

`start` gets the index of the first character that is not a space: only once, while `start` is still `-1`. `end` gets the index of each character that is not a space, so after the loop it is the index of the last one. If the string has only spaces, `start` stays `-1`, and the function returns `''`. The slice is `s[start:end + 1]`, because a slice stops before its second index.

### Question 37 (hard): write a function

Write a function called `sort_list` that sorts a list of numbers from the smallest to the largest. Do not use the function `sorted` or the list method `sort`. Change the list itself, and return it too.

Use a nested loop: for each index `i`, compare `lst[i]` with each item after it, `lst[j]`. If `lst[j]` is smaller, swap the two items. When the inner loop is done, `lst[i]` holds the smallest item of `lst[i:]`.

```python
lst[j] = temp
```

```python
    ...
```

**Examples**


```text
[1, 2, 4, 8]
```

```python
sort_list([3, 1, 2, 1])
```

```text
[1, 1, 2, 3]
```

```python
numbers
```

```text
[3, 4, 5]
```

:::{admonition} Answer
:class: dropdown

Solution 1 (swap each time a later item is smaller):

```python
def sort_list(lst):
    for i in range(len(lst)):
        for j in range(i + 1, len(lst)):
            if lst[j] < lst[i]:
                temp = lst[i]
                lst[i] = lst[j]
                lst[j] = temp
    return lst
```
Then try: `sort_list([4, 2, 8, 1])`.

Solution 2 (selection sort: find the index of the smallest item, then swap only once):

```python
def sort_list(lst):
    for i in range(len(lst)):
        smallest = i
        for j in range(i + 1, len(lst)):
            if lst[j] < lst[smallest]:
                smallest = j
        temp = lst[i]
        lst[i] = lst[smallest]
        lst[smallest] = temp
    return lst
```

Both functions change the list that they get. In the third example, `numbers` and `lst` refer to the same list, so after the call, `numbers` is sorted too.

To return a new list, sort a copy: start the function with `lst = lst[:]` (a slice of the whole list is a new list). Then the caller's list does not change.

```python
def count_items(lst):
    uniques = []
    for x in lst:
        if x not in uniques:
            uniques.append(x)
    result = []
    for x in sorted(uniques):
        count = 0
        for y in lst:
            if y == x:
                count = count + 1
        result.append([x, count])
    return result
```
